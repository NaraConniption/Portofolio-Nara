```python
# ═══════════════════════════════════════════════════════════════════════════════
#  RNN vs LSTM vs GRU  —  6 Targets
#  XAI : Permutation Importance  +  SHAP (GradientExplainer)
#          ↳ SHAP Bar  |  SHAP Beeswarm  |  SHAP Temporal Heatmap
#          ↳ SHAP Cross-Model Comparison  |  SHAP + Metrics Joint Dashboard
#  + Anomaly Score  +  Sequence Prediction Accuracy
#  + Composite Model Selection Dashboard
# ═══════════════════════════════════════════════════════════════════════════════

import os, sys, subprocess
os.environ['CUDA_VISIBLE_DEVICES'] = '-1'
os.environ['TF_CPP_MIN_LOG_LEVEL'] = '3'

# ── Auto-install SHAP if missing ──────────────────────────────────────────────
try:
    import shap
    print(f"✅ SHAP {shap.__version__} already installed")
except ImportError:
    print("📦 Installing SHAP …")
    subprocess.check_call([sys.executable, '-m', 'pip', 'install', 'shap', '-q'])
    import shap
    print(f"✅ SHAP {shap.__version__} installed")

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.dates as mdates
import matplotlib.gridspec as gridspec
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')

from sklearn.preprocessing import MinMaxScaler
from sklearn.metrics import mean_squared_error, r2_score

# ── TensorFlow ────────────────────────────────────────────────────────────────
try:
    import tensorflow as tf
    from tensorflow.keras.models import Sequential
    from tensorflow.keras.layers import SimpleRNN, LSTM, GRU, Dense, Dropout
    from tensorflow.keras.optimizers import Adam
    from tensorflow.keras.callbacks import EarlyStopping
    print(f"✅ TensorFlow {tf.__version__} OK")
except Exception as e:
    print(f"❌ TensorFlow error: {e}"); sys.exit()

# ── White theme ───────────────────────────────────────────────────────────────
plt.style.use('default')
plt.rcParams.update({
    'figure.facecolor': 'white', 'axes.facecolor': 'white',
    'axes.edgecolor': '#cccccc', 'axes.labelcolor': '#222222',
    'xtick.color': '#444444', 'ytick.color': '#444444',
    'text.color': '#222222', 'legend.facecolor': 'white',
    'legend.edgecolor': '#cccccc', 'grid.color': '#eeeeee',
    'grid.alpha': 0.6, 'font.size': 10,
})

print("=" * 80)
print("  RNN vs LSTM vs GRU  |  Perm-XAI + SHAP + Anomaly + SeqAcc")
print("=" * 80)

# ─────────────────────────────────────────────────────────────────────────────
# CONFIGURATION
# ─────────────────────────────────────────────────────────────────────────────
MODEL_CONFIGS = {
    'RNN':  {'seq': 32, 'units': 128, 'color': '#e63946', 'marker': 'o'},
    'LSTM': {'seq': 32, 'units': 128, 'color': '#2a9d8f', 'marker': 's'},
    'GRU':  {'seq': 32, 'units': 256, 'color': '#f4a261', 'marker': '^'},
}

TRAIN_CFG = {
    'epochs': 300, 'batch_size': 32, 'dropout': 0.2,
    'system_capacity_wp': 2250, 'panel_area_m2': 10.893,
}

TARGETS = [
    'Power PV', 'PR (%)', 'Efficiency PV (%)',
    'Efficiency Temp (%)', 'Efficiency Battery (%)', 'Efficiency Inv (%)',
]

LEAKAGE_MAP = {
    'Power PV':               ['Power PV', 'Efficiency PV (%)',
                               'Efficiency System (%)', 'PR (%)', 'P_pv'],
    'PR (%)':                 ['PR (%)'],
}

# XAI / quality thresholds
XAI_N_PERM           = 10     # permutation repeats per feature
SHAP_BG              = 80     # background samples for GradientExplainer
SHAP_EXPLAIN         = 100    # test samples to explain
SHAP_TOP_N           = 12     # features shown in SHAP plots
ANOMALY_THRESH       = 2.5    # z-score threshold
SEQ_TOL              = 0.10   # ±10 % tolerance band


# ─────────────────────────────────────────────────────────────────────────────
# 1.  DATA LOADING
# ─────────────────────────────────────────────────────────────────────────────
def load_hybrid_data(env_file='/content/Lingkungan.csv',
                     dc_file='/content/Elektrik DC.csv',
                     ac_file='/content/Elektrik AC.csv'):
    print("\n📁 Loading 3 CSV files …")
    df_env = pd.read_csv(env_file)
    df_dc  = pd.read_csv(dc_file)
    df_ac  = pd.read_csv(ac_file)

    def parse_dt(df):
        df = df.copy()
        df['datetime'] = pd.to_datetime(
            df['Date'].astype(str) + ' ' + df['Time'].astype(str), errors='coerce')
        return df.dropna(subset=['datetime']).sort_values('datetime')

    df_env = parse_dt(df_env).set_index('datetime')
    df_dc  = parse_dt(df_dc ).set_index('datetime')
    df_ac  = parse_dt(df_ac ).set_index('datetime')

    tol = pd.Timedelta('10m')
    df  = pd.merge_asof(df_dc.sort_index(), df_env.sort_index(),
                        left_index=True, right_index=True,
                        tolerance=tol, direction='nearest')
    df  = pd.merge_asof(df.sort_index(), df_ac.sort_index(),
                        left_index=True, right_index=True,
                        tolerance=tol, direction='nearest')

    df['hour']       = df.index.hour
    df['is_daytime'] = ((df.index.hour >= 6) & (df.index.hour <= 18)).astype(int)
    df['hour_sin']   = np.sin(2 * np.pi * df.index.hour / 24)
    df['hour_cos']   = np.cos(2 * np.pi * df.index.hour / 24)

    print(f"✅ Merged: {len(df)} rows, {df.shape[1]} columns")
    return df


# ─────────────────────────────────────────────────────────────────────────────
# 2.  FEATURE ENGINEERING
# ─────────────────────────────────────────────────────────────────────────────
def engineer_features(df, system_capacity_wp=2250, panel_area_m2=10.893):
    df = df.copy()
    for col in df.columns:
        df[col] = pd.to_numeric(df[col], errors='coerce')

    df['delta_t'] = df.index.to_series().diff().dt.total_seconds() / 3600
    df['P_ac']    = df['V AC'] * df['I AC']
    df['E_ac']    = df['P_ac'] * df['delta_t']

    if 'Irradiance' in df.columns and 'Power PV' in df.columns:
        df['Efficiency PV (%)'] = (
            df['Power PV'] / (df['Irradiance'] * panel_area_m2)) * 100

    df['P_batt']      = df['V25 (V)'] * df['I ACS758 (A)']
    df['E_batt']      = df['P_batt'] * df['delta_t']
    df['is_charging'] = (
        (df.index.hour >= 6) & (df.index.hour < 18) &
        (df['V25 (V)'].fillna(0) > 0) &
        (df['I ACS758 (A)'].fillna(0) > 0)
    )
    df['E_charge']    = np.where(df['is_charging'],  df['E_batt'], np.nan)
    df['E_discharge'] = np.where(~df['is_charging'], df['E_batt'], np.nan)

    df['E_charge_total']    = (df['E_charge'].fillna(0)
                                .rolling('24h', min_periods=1).sum().shift(1))
    df['E_discharge_total'] = (df['E_discharge'].fillna(0)
                                .rolling('24h', min_periods=1).sum().shift(1))
    df['Efficiency Battery (%)'] = (
        df['E_discharge_total'] / df['E_charge_total'].replace(0, np.nan)) * 100

    df['P_pv']           = df['V220 (V)'] * df['I ACS712 (A)']
    df['E_dc_total']     = df['P_pv'] * df['delta_t']
    df['E_dc_total_24h'] = (df['E_dc_total']
                             .rolling('24h', min_periods=1).sum().shift(1))
    df['E_ac_24h']       = df['E_ac'].rolling('24h', min_periods=1).sum().shift(1)
    df['Efficiency Inv (%)'] = (df['E_ac_24h'] / df['E_dc_total_24h']) * 100
    df['Efficiency Inv (%)'] = df['Efficiency Inv (%)'].replace([np.inf, -np.inf], np.nan)

    df['Efficiency Temp (%)'] = (
        1 - 0.004 * (df['Temp PV (C)'].fillna(25) - 25)) * 100

    df['Irrad_kWh']       = (df['Irradiance'] * df['delta_t']) / 1000
    df['Irrad_total_24h'] = (df['Irrad_kWh']
                              .rolling('24h', min_periods=1).sum().shift(1))
    df['PR (%)'] = (
        df['E_ac_24h'] / (system_capacity_wp * df['Irrad_total_24h'])) * 100

    df['Efficiency PV (%)']       = df['Efficiency PV (%)'].clip(0, 25)
    df['Efficiency Battery (%)']  = df['Efficiency Battery (%)'].clip(0, 100)
    df['Efficiency Inv (%)']      = df['Efficiency Inv (%)'].clip(0, 100)
    df['Efficiency Temp (%)']     = df['Efficiency Temp (%)'].clip(50, 100)
    df['PR (%)']                  = df['PR (%)'].clip(0, 100)

    print("✅ Feature engineering done — 6 targets ready")
    return df


def clean_data(df):
    df = df.copy()
    for col in df.select_dtypes(include=[np.number]).columns:
        df[col] = df[col].fillna(df[col].median())
        df[col] = np.nan_to_num(df[col])
        q1, q3 = df[col].quantile([0.25, 0.75])
        df[col] = df[col].clip(q1 - 1.5*(q3-q1), q3 + 1.5*(q3-q1))
    return df


# ─────────────────────────────────────────────────────────────────────────────
# 3.  BUILD MODEL
# ─────────────────────────────────────────────────────────────────────────────
def build_model(arch, seq_len, n_features, units, dropout=0.2):
    m = Sequential(name=arch)
    if arch == 'RNN':
        m.add(SimpleRNN(units, input_shape=(seq_len, n_features),
                        return_sequences=False))
    elif arch == 'LSTM':
        m.add(LSTM(units, input_shape=(seq_len, n_features),
                   return_sequences=False))
    elif arch == 'GRU':
        m.add(GRU(units, input_shape=(seq_len, n_features),
                  return_sequences=False))
    m.add(Dropout(dropout))
    m.add(Dense(32, activation='relu'))
    m.add(Dense(1))
    m.compile(optimizer=Adam(learning_rate=0.001), loss='mse')
    return m


# ─────────────────────────────────────────────────────────────────────────────
# 4A.  XAI — PERMUTATION FEATURE IMPORTANCE
# ─────────────────────────────────────────────────────────────────────────────
def compute_permutation_importance(model, Xte, y_test_orig, sc_y,
                                   feature_names, n_repeats=XAI_N_PERM):
    y_base   = sc_y.inverse_transform(model.predict(Xte, verbose=0)).flatten()
    base_mse = mean_squared_error(y_test_orig, y_base)

    imp = np.zeros(len(feature_names))
    for fi in range(len(feature_names)):
        deltas = []
        for _ in range(n_repeats):
            Xp = Xte.copy()
            Xp[:, :, fi] = Xp[np.random.permutation(len(Xp)), :, fi]
            yp = sc_y.inverse_transform(model.predict(Xp, verbose=0)).flatten()
            deltas.append(mean_squared_error(y_test_orig, yp) - base_mse)
        imp[fi] = np.mean(deltas)

    imp = np.clip(imp, 0, None)
    tot = imp.sum()
    if tot > 0:
        imp /= tot
    return imp


# ─────────────────────────────────────────────────────────────────────────────
# 4B.  XAI — SHAP  (GradientExplainer, fallback KernelExplainer)
#
#  Input tensor shape : (samples, seq_len, n_features)
#  Raw SHAP values    : (samples, seq_len, n_features)  — signed
#  shap_2d            : mean(|SHAP|) over seq → (samples, n_features)
#  shap_3d            : raw signed values     → (samples, seq_len, n_features)
#  X_exp_2d           : mean feature values over seq → (samples, n_features)
# ─────────────────────────────────────────────────────────────────────────────
def compute_shap(model, Xtr, Xte, feature_names,
                 bg=SHAP_BG, n_explain=SHAP_EXPLAIN):
    print(f"      🔮 SHAP GradientExplainer  bg={bg}  explain={n_explain} …")

    bg_idx  = np.random.choice(len(Xtr), size=min(bg, len(Xtr)), replace=False)
    X_bg    = Xtr[bg_idx].astype(np.float32)
    ex_idx  = np.random.choice(len(Xte), size=min(n_explain, len(Xte)), replace=False)
    X_exp   = Xte[ex_idx].astype(np.float32)

    try:
        explainer = shap.GradientExplainer(model, X_bg)
        sv        = np.array(explainer.shap_values(X_exp))

        # handle list output (multi-output models)
        if isinstance(sv, list):
            sv = sv[0]
        # squeeze trailing 1-dims: e.g. (S, seq, F, 1) → (S, seq, F)
        while sv.ndim > 3:
            sv = sv.squeeze(-1)

        shap_3d  = sv                                    # (S, seq, F)
        shap_2d  = np.abs(sv).mean(axis=1)              # (S, F)
        X_exp_2d = X_exp.mean(axis=1)                   # (S, F)

        print(f"         ✅ GradientExplainer done — shape {shap_2d.shape}")
        return shap_2d, shap_3d, X_exp_2d

    except Exception as e:
        print(f"         ⚠️  GradientExplainer failed ({e})")
        print("         ↳  Trying KernelExplainer (slower) …")
        try:
            n_s, seq, nf = X_bg.shape
            X_bg_2d  = X_bg.reshape(n_s, seq * nf)
            X_exp_2d_flat = X_exp.reshape(len(X_exp), seq * nf)

            def model_fn(x2d):
                return model.predict(
                    x2d.reshape(-1, seq, nf).astype(np.float32),
                    verbose=0).flatten()

            ke  = shap.KernelExplainer(model_fn, shap.sample(X_bg_2d, 30))
            sv2 = ke.shap_values(X_exp_2d_flat, nsamples=50, silent=True)
            sv3 = np.array(sv2).reshape(-1, seq, nf)
            s2d = np.abs(sv3).mean(axis=1)
            x2d = X_exp.mean(axis=1)
            print(f"         ✅ KernelExplainer done — shape {s2d.shape}")
            return s2d, sv3, x2d
        except Exception as e2:
            print(f"         ❌ KernelExplainer also failed: {e2}")
            return None, None, None


# ─────────────────────────────────────────────────────────────────────────────
# 5.  ANOMALY SCORE
# ─────────────────────────────────────────────────────────────────────────────
def anomaly_score(y_actual, y_pred, thresh=ANOMALY_THRESH):
    res   = y_actual - y_pred
    mu, sigma = res.mean(), res.std()
    if sigma == 0:
        sigma = 1e-9
    z     = (res - mu) / sigma
    flags = np.abs(z) > thresh
    return {
        'flags':           flags,
        'z_scores':        z,
        'anomaly_rate':    flags.mean() * 100,
        'mean_abs_zscore': np.abs(z).mean(),
        'n_anomalies':     int(flags.sum()),
    }


# ─────────────────────────────────────────────────────────────────────────────
# 6.  SEQUENCE PREDICTION ACCURACY
# ─────────────────────────────────────────────────────────────────────────────
def seq_accuracy(y_actual, y_pred, tol_pct=SEQ_TOL):
    tol    = np.maximum(np.abs(y_actual) * tol_pct, 1e-6)
    tba    = (np.abs(y_actual - y_pred) <= tol).mean() * 100
    da     = (np.sign(np.diff(y_actual)) == np.sign(np.diff(y_pred))).mean() * 100
    return {'tba': tba, 'da': da, 'seq_score': (tba + da) / 2}


# ─────────────────────────────────────────────────────────────────────────────
# 7.  TRAIN ONE ARCH × ONE TARGET
# ─────────────────────────────────────────────────────────────────────────────
def train_model(df, target_col, arch):
    cfg     = MODEL_CONFIGS[arch]
    seq_len = cfg['seq']

    all_num    = df.select_dtypes('number').columns.tolist()
    leak       = LEAKAGE_MAP.get(target_col, [target_col])
    features_X = [c for c in all_num if c not in leak]

    if target_col not in df.columns or len(features_X) < 3:
        print(f"   ⚠️  {arch}/{target_col}: skipped"); return None

    X_raw = df[features_X].ffill().values
    y_raw = df[target_col].ffill().values
    ts    = df.index[seq_len:]

    X_seq, y_seq = [], []
    for i in range(seq_len, len(X_raw)):
        X_seq.append(X_raw[i-seq_len:i])
        y_seq.append(y_raw[i])
    if len(X_seq) < 100:
        print(f"   ⚠️  {arch}/{target_col}: not enough sequences"); return None

    X_arr = np.array(X_seq, dtype=np.float32)
    y_arr = np.array(y_seq, dtype=np.float32)
    split = int(0.9 * len(X_arr))
    Xtr, Xte = X_arr[:split], X_arr[split:]
    ytr, yte = y_arr[:split], y_arr[split:]
    test_ts   = ts[split:]

    nf   = Xtr.shape[2]
    sc_X = MinMaxScaler(); sc_y = MinMaxScaler()
    Xtr  = sc_X.fit_transform(Xtr.reshape(-1, nf)).reshape(Xtr.shape)
    Xte  = sc_X.transform(Xte.reshape(-1, nf)).reshape(Xte.shape)
    ytr  = sc_y.fit_transform(ytr.reshape(-1, 1)).flatten()
    yte  = sc_y.transform(yte.reshape(-1, 1)).flatten()

    model = build_model(arch, seq_len, nf, cfg['units'], TRAIN_CFG['dropout'])
    model.fit(Xtr, ytr,
              epochs=TRAIN_CFG['epochs'],
              batch_size=TRAIN_CFG['batch_size'],
              validation_split=0.1,
              callbacks=[EarlyStopping(patience=20, restore_best_weights=True)],
              verbose=0)

    y_pred      = sc_y.inverse_transform(
                    model.predict(Xte, verbose=0)).flatten()
    y_test_orig = sc_y.inverse_transform(yte.reshape(-1, 1)).flatten()

    # ── core metrics ──────────────────────────────────────────────────────────
    rmse  = np.sqrt(mean_squared_error(y_test_orig, y_pred))
    r2    = r2_score(y_test_orig, y_pred)
    ymax  = np.max(np.abs(y_test_orig))
    nrmse = (rmse / ymax * 100) if ymax > 0 else np.nan

    # ── XAI-1: Permutation ────────────────────────────────────────────────────
    print(f"      🔍 Permutation importance ({arch}/{target_col}) …")
    perm_imp = compute_permutation_importance(
        model, Xte, y_test_orig, sc_y, features_X)

    # ── XAI-2: SHAP ───────────────────────────────────────────────────────────
    shap_2d, shap_3d, X_exp_2d = compute_shap(model, Xtr, Xte, features_X)

    # ── anomaly + seq acc ─────────────────────────────────────────────────────
    anom  = anomaly_score(y_test_orig, y_pred)
    s_acc = seq_accuracy(y_test_orig, y_pred)

    print(f"   ✅ {arch}/{target_col}  "
          f"nRMSE={nrmse:.1f}%  R²={r2:.3f}  "
          f"Anomaly={anom['anomaly_rate']:.1f}%  "
          f"SeqAcc={s_acc['seq_score']:.1f}%")

    return {
        'arch': arch, 'target': target_col,
        'model': model, 'feature_names': features_X,
        'scaler_x': sc_X, 'scaler_y': sc_y,
        'X_train_scaled': Xtr, 'X_test_scaled': Xte,
        'y_test_orig': y_test_orig, 'y_pred': y_pred,
        'test_timestamps': test_ts,
        # core
        'rmse': rmse, 'nrmse': nrmse, 'r2': r2,
        # XAI
        'perm_imp': perm_imp,
        'shap_2d': shap_2d, 'shap_3d': shap_3d, 'X_exp_2d': X_exp_2d,
        # anomaly
        'anomaly': anom,
        # seq acc
        'seq_acc': s_acc,
    }


# ─────────────────────────────────────────────────────────────────────────────
# 8.  COMPOSITE SCORE  (lower = better)
# ─────────────────────────────────────────────────────────────────────────────
def composite(res):
    nr  = res['nrmse'] if not np.isnan(res['nrmse']) else 100.0
    r2p = (1 - res['r2']) * 100
    an  = res['anomaly']['anomaly_rate']
    sp  = 100 - res['seq_acc']['seq_score']
    return 0.35*nr + 0.25*r2p + 0.25*an + 0.15*sp


# ═══════════════════════════════════════════════════════════════════════════════
# 9.  PLOT FUNCTIONS
# ═══════════════════════════════════════════════════════════════════════════════

# ── 9A: Actual vs Prediction + anomaly markers ────────────────────────────────
def plot_predictions(rba):
    for arch, tr in rba.items():
        cfg   = MODEL_CONFIGS[arch]
        valid = {t: r for t, r in tr.items() if r}
        if not valid: continue

        fig, axes = plt.subplots(3, 2, figsize=(22, 18))
        fig.patch.set_facecolor('white')
        axes = axes.flatten()
        fig.suptitle(f'{arch} — Actual vs Prediction  |  6 Targets',
                     fontsize=18, fontweight='bold')

        for idx, (tgt, res) in enumerate(valid.items()):
            ax = axes[idx]
            t  = pd.to_datetime(res['test_timestamps'])
            a, p = res['y_test_orig'], res['y_pred']

            ax.plot(t, a, color='#4361ee', lw=2,   label='Actual',    alpha=0.9)
            ax.plot(t, p, color=cfg['color'], lw=2, label='Prediction',
                    ls='--', alpha=0.9)
            ax.fill_between(t, a, p, alpha=0.15, color=cfg['color'])

            fl = res['anomaly']['flags']
            ax.scatter(t[fl], a[fl], color='red', s=22, zorder=5,
                       label=f"Anomaly ({fl.sum()})")

            ax.set_title(
                f"{tgt}\nR²={res['r2']:.3f}  "
                f"nRMSE={res['nrmse']:.1f}%  "
                f"SeqAcc={res['seq_acc']['seq_score']:.1f}%",
                fontsize=9, fontweight='bold')
            ax.xaxis.set_major_formatter(mdates.DateFormatter('%d-%b\n%H:%M'))
            ax.legend(fontsize=8); ax.grid(True, alpha=0.4)

        plt.tight_layout()
        plt.savefig(f'/content/{arch}_predictions.png', dpi=150, bbox_inches='tight')
        plt.show()
        print(f"✅ Saved: {arch}_predictions.png")


# ── 9B: Permutation importance ────────────────────────────────────────────────
def plot_perm_xai(rba, top_n=10):
    for tgt in TARGETS:
        avail = {a: rba[a].get(tgt) for a in MODEL_CONFIGS if rba[a].get(tgt)}
        if not avail: continue

        n = len(avail)
        fig, axes = plt.subplots(1, n, figsize=(8*n, 6))
        if n == 1: axes = [axes]
        fig.suptitle(f'Permutation Feature Importance  |  {tgt}',
                     fontsize=14, fontweight='bold')

        for ax, (arch, res) in zip(axes, avail.items()):
            imp   = res['perm_imp']
            names = np.array(res['feature_names'])
            order = np.argsort(imp)[::-1][:top_n]
            bars  = ax.barh(names[order][::-1], imp[order][::-1],
                            color=MODEL_CONFIGS[arch]['color'], alpha=0.85)
            for b in bars:
                w = b.get_width()
                ax.text(w+0.002, b.get_y()+b.get_height()/2,
                        f'{w:.3f}', va='center', fontsize=8)
            ax.set_title(f'{arch}  (top {top_n})', fontweight='bold')
            ax.set_xlabel('Normalised ΔMSE')
            ax.grid(axis='x', alpha=0.4)

        plt.tight_layout()
        safe = tgt.replace(' ','_').replace('/','_').replace('%','pct')
        plt.savefig(f'/content/Perm_XAI_{safe}.png', dpi=150, bbox_inches='tight')
        plt.show()
        print(f"✅ Saved: Perm_XAI_{safe}.png")


# ── 9C: SHAP Bar (mean |SHAP value| per feature) ─────────────────────────────
def plot_shap_bar(rba, top_n=SHAP_TOP_N):
    for tgt in TARGETS:
        avail = {a: rba[a].get(tgt) for a in MODEL_CONFIGS
                 if rba[a].get(tgt) and rba[a][tgt].get('shap_2d') is not None}
        if not avail:
            print(f"   ⚠️  SHAP bar: no data for {tgt}"); continue

        n = len(avail)
        fig, axes = plt.subplots(1, n, figsize=(8*n, 6))
        if n == 1: axes = [axes]
        fig.suptitle(f'SHAP  mean |value|  — Feature Importance  |  {tgt}',
                     fontsize=14, fontweight='bold')

        for ax, (arch, res) in zip(axes, avail.items()):
            sv2   = res['shap_2d']                     # (S, F)
            names = np.array(res['feature_names'])
            m_abs = sv2.mean(axis=0)                   # (F,)
            order = np.argsort(m_abs)[::-1][:top_n]

            pal   = plt.cm.RdYlGn_r(np.linspace(0.1, 0.9, top_n))
            bars  = ax.barh(names[order][::-1], m_abs[order][::-1],
                            color=pal[::-1], alpha=0.9, edgecolor='white')
            for b in bars:
                w = b.get_width()
                ax.text(w+2e-4, b.get_y()+b.get_height()/2,
                        f'{w:.4f}', va='center', fontsize=8)

            ax.set_title(arch, fontweight='bold',
                         color=MODEL_CONFIGS[arch]['color'])
            ax.set_xlabel('mean(|SHAP value|)')
            ax.axvline(0, color='black', lw=0.8)
            ax.grid(axis='x', alpha=0.35)

        plt.tight_layout()
        safe = tgt.replace(' ','_').replace('/','_').replace('%','pct')
        plt.savefig(f'/content/SHAP_bar_{safe}.png', dpi=150, bbox_inches='tight')
        plt.show()
        print(f"✅ Saved: SHAP_bar_{safe}.png")


# ── 9D: SHAP Beeswarm (signed, dots coloured by feature value) ───────────────
def plot_shap_beeswarm(rba, top_n=SHAP_TOP_N):
    for arch in MODEL_CONFIGS:
        for tgt in TARGETS:
            res = rba[arch].get(tgt)
            if res is None or res.get('shap_3d') is None: continue

            sv3    = res['shap_3d']            # (S, seq, F)
            X_2d   = res['X_exp_2d']           # (S, F) — mean over seq
            names  = np.array(res['feature_names'])

            sv_sig = sv3.mean(axis=1)          # (S, F) signed
            m_abs  = np.abs(sv_sig).mean(axis=0)
            order  = np.argsort(m_abs)[::-1][:top_n]
            ordr   = order[::-1]               # bottom→top
            nf     = len(ordr)

            fig, ax = plt.subplots(figsize=(10, max(5, nf*0.55)))
            fig.patch.set_facecolor('white')
            cmap = plt.cm.coolwarm

            for ri, fi in enumerate(ordr):
                sv_f = sv_sig[:, fi]
                xf   = X_2d[:, fi]
                xf_n = (xf - xf.min()) / (np.ptp(xf) + 1e-9)
                colors = cmap(xf_n)
                jitter = np.random.uniform(-0.3, 0.3, size=len(sv_f))
                ax.scatter(sv_f, ri + jitter, c=colors, s=12, alpha=0.7,
                           linewidths=0, zorder=3)

            ax.set_yticks(range(nf))
            ax.set_yticklabels(names[ordr], fontsize=9)
            ax.axvline(0, color='black', lw=1, zorder=4)
            ax.set_xlabel('SHAP value  (impact on model output)', fontsize=10)
            ax.set_title(f'SHAP Beeswarm  |  {arch}  —  {tgt}',
                         fontsize=12, fontweight='bold',
                         color=MODEL_CONFIGS[arch]['color'])
            ax.grid(axis='x', alpha=0.35)

            sm = plt.cm.ScalarMappable(cmap=cmap,
                                        norm=plt.Normalize(vmin=0, vmax=1))
            sm.set_array([])
            cb = plt.colorbar(sm, ax=ax, shrink=0.5, pad=0.01)
            cb.set_label('Feature value\n(low → high)', fontsize=8)

            plt.tight_layout()
            safe_t = tgt.replace(' ','_').replace('/','_').replace('%','pct')
            fname  = f'/content/SHAP_beeswarm_{arch}_{safe_t}.png'
            plt.savefig(fname, dpi=150, bbox_inches='tight')
            plt.show()
            print(f"✅ Saved: SHAP_beeswarm_{arch}_{safe_t}.png")


# ── 9E: SHAP Temporal Heatmap (seq-step × feature) ───────────────────────────
def plot_shap_temporal(rba, top_n=8):
    """
    Rows = features, Cols = sequence timesteps.
    Darker = higher |SHAP| at that lag position.
    Reveals which time-lags each model relies on.
    """
    for arch in MODEL_CONFIGS:
        for tgt in TARGETS:
            res = rba[arch].get(tgt)
            if res is None or res.get('shap_3d') is None: continue

            sv3   = res['shap_3d']               # (S, seq, F)
            names = np.array(res['feature_names'])

            heat  = np.abs(sv3).mean(axis=0)     # (seq, F)
            top_i = np.argsort(heat.mean(axis=0))[::-1][:top_n]
            heat_t = heat[:, top_i]              # (seq, top_n)

            fig, ax = plt.subplots(figsize=(14, max(4, top_n*0.6)))
            fig.patch.set_facecolor('white')
            im = ax.imshow(heat_t.T, aspect='auto', cmap='YlOrRd', origin='lower')

            ax.set_yticks(range(top_n))
            ax.set_yticklabels(names[top_i], fontsize=9)
            ax.set_xlabel(
                'Sequence step  (0 = oldest  →  right = most recent)',
                fontsize=10)
            ax.set_title(
                f'SHAP Temporal Heatmap  |  {arch}  —  {tgt}\n'
                f'mean |SHAP| per timestep × feature  '
                f'(darker = more influential lag)',
                fontsize=11, fontweight='bold',
                color=MODEL_CONFIGS[arch]['color'])
            plt.colorbar(im, ax=ax, label='mean |SHAP|', shrink=0.8)
            plt.tight_layout()

            safe_t = tgt.replace(' ','_').replace('/','_').replace('%','pct')
            fname  = f'/content/SHAP_temporal_{arch}_{safe_t}.png'
            plt.savefig(fname, dpi=150, bbox_inches='tight')
            plt.show()
            print(f"✅ Saved: SHAP_temporal_{arch}_{safe_t}.png")


# ── 9F: SHAP Cross-Model (which features 3 models agree on) ──────────────────
def plot_shap_cross_model(rba, top_n=10):
    arch_list = list(MODEL_CONFIGS.keys())

    for tgt in TARGETS:
        avail = {a: rba[a].get(tgt) for a in arch_list
                 if rba[a].get(tgt) and rba[a][tgt].get('shap_2d') is not None}
        if not avail: continue

        # union top features
        union_mean = {}
        for arch, res in avail.items():
            mv = res['shap_2d'].mean(axis=0)
            for i, nm in enumerate(res['feature_names']):
                union_mean[nm] = union_mean.get(nm, 0) + mv[i]

        top_feats = sorted(union_mean, key=union_mean.get, reverse=True)[:top_n]

        fig, ax = plt.subplots(figsize=(14, 6))
        fig.patch.set_facecolor('white')
        x   = np.arange(len(top_feats))
        na  = len(avail)
        w   = 0.8 / na
        off = np.linspace(-(na-1)*w/2, (na-1)*w/2, na)

        for i, (arch, res) in enumerate(avail.items()):
            mv    = res['shap_2d'].mean(axis=0)
            fn    = res['feature_names']
            vals  = [mv[fn.index(f)] if f in fn else 0 for f in top_feats]
            bars  = ax.bar(x + off[i], vals, w,
                           label=arch, color=MODEL_CONFIGS[arch]['color'],
                           alpha=0.85, edgecolor='white')
            for b in bars:
                h = b.get_height()
                if h > 0:
                    ax.text(b.get_x()+b.get_width()/2, h+3e-4,
                            f'{h:.3f}', ha='center', va='bottom', fontsize=6.5)

        ax.set_xticks(x)
        ax.set_xticklabels(top_feats, rotation=35, ha='right', fontsize=9)
        ax.set_ylabel('mean |SHAP value|')
        ax.set_title(f'SHAP Cross-Model Comparison  |  {tgt}',
                     fontsize=13, fontweight='bold')
        ax.legend(fontsize=10); ax.grid(axis='y', alpha=0.35)
        plt.tight_layout()

        safe_t = tgt.replace(' ','_').replace('/','_').replace('%','pct')
        plt.savefig(f'/content/SHAP_crossmodel_{safe_t}.png', dpi=150, bbox_inches='tight')
        plt.show()
        print(f"✅ Saved: SHAP_crossmodel_{safe_t}.png")


# ── 9G: Anomaly dashboard ─────────────────────────────────────────────────────
def plot_anomaly_dashboard(rba):
    rows, cols = len(TARGETS), len(MODEL_CONFIGS)
    fig, axes  = plt.subplots(rows, cols, figsize=(7*cols, 4*rows))
    fig.patch.set_facecolor('white')
    fig.suptitle('Anomaly Score Dashboard  |  Z-score Distribution',
                 fontsize=16, fontweight='bold', y=1.01)

    arch_list = list(MODEL_CONFIGS.keys())
    for r, tgt in enumerate(TARGETS):
        for c, arch in enumerate(arch_list):
            ax  = axes[r][c]
            res = rba[arch].get(tgt)
            if res is None:
                ax.set_visible(False); continue

            z     = res['anomaly']['z_scores']
            flags = res['anomaly']['flags']
            ax.hist(z[~flags], bins=30, color=MODEL_CONFIGS[arch]['color'],
                    alpha=0.7, label='Normal')
            if flags.sum() > 0:
                ax.hist(z[flags], bins=10, color='red', alpha=0.85,
                        label=f"Anomaly ({flags.sum()})")
            ax.axvline(-ANOMALY_THRESH, color='red', ls='--', lw=1.2)
            ax.axvline( ANOMALY_THRESH, color='red', ls='--', lw=1.2)
            ax.set_title(
                f'{arch} | {tgt}\n'
                f'Rate={res["anomaly"]["anomaly_rate"]:.1f}%  '
                f'AvgZ={res["anomaly"]["mean_abs_zscore"]:.2f}',
                fontsize=8, fontweight='bold')
            ax.legend(fontsize=7); ax.grid(True, alpha=0.3)
            if c == 0: ax.set_ylabel('Count', fontsize=8)

    plt.tight_layout()
    plt.savefig('/content/anomaly_dashboard.png', dpi=150, bbox_inches='tight')
    plt.show()
    print("✅ Saved: anomaly_dashboard.png")


# ── 9H: Sequence Accuracy comparison ─────────────────────────────────────────
def plot_seq_accuracy(rba):
    arch_list = list(MODEL_CONFIGS.keys())
    fig, axes = plt.subplots(2, 3, figsize=(21, 10))
    fig.patch.set_facecolor('white')
    fig.suptitle('Sequence Prediction Accuracy  |  TBA & DA per Target',
                 fontsize=16, fontweight='bold')
    axes = axes.flatten()

    for idx, tgt in enumerate(TARGETS):
        ax = axes[idx]
        x  = np.arange(len(arch_list))
        w  = 0.25
        tba_v = [rba[a].get(tgt, {}) and rba[a][tgt] and
                 rba[a][tgt]['seq_acc']['tba'] or 0 for a in arch_list]
        da_v  = [rba[a].get(tgt, {}) and rba[a][tgt] and
                 rba[a][tgt]['seq_acc']['da']  or 0 for a in arch_list]

        def _get(a, key):
            r = rba[a].get(tgt)
            return r['seq_acc'][key] if r else 0

        tba_v = [_get(a, 'tba') for a in arch_list]
        da_v  = [_get(a, 'da')  for a in arch_list]

        b1 = ax.bar(x-w/2, tba_v, w, label='TBA (±10%)',
                    color=[MODEL_CONFIGS[a]['color'] for a in arch_list], alpha=0.85)
        b2 = ax.bar(x+w/2, da_v, w, label='Directional Acc',
                    color=[MODEL_CONFIGS[a]['color'] for a in arch_list],
                    alpha=0.45, hatch='//')
        ax.set_xticks(x); ax.set_xticklabels(arch_list, fontweight='bold')
        ax.set_ylim(0, 115)
        ax.axhline(80, color='green', ls='--', lw=1, label='80% target')
        ax.set_title(tgt, fontweight='bold')
        ax.set_ylabel('Accuracy (%)')
        ax.legend(fontsize=7); ax.grid(axis='y', alpha=0.4)
        for b in list(b1)+list(b2):
            h = b.get_height()
            if h > 0:
                ax.text(b.get_x()+b.get_width()/2, h+0.5,
                        f'{h:.1f}', ha='center', fontsize=7)

    plt.tight_layout()
    plt.savefig('/content/seq_accuracy.png', dpi=150, bbox_inches='tight')
    plt.show()
    print("✅ Saved: seq_accuracy.png")


# ── 9I: Model Selection Dashboard (composite heatmap + 4 metric bars) ─────────
def plot_model_selection_dashboard(rba):
    arch_list = list(MODEL_CONFIGS.keys())
    tgt_list  = TARGETS

    records = []
    for arch in arch_list:
        for tgt in tgt_list:
            res = rba[arch].get(tgt)
            if res is None:
                records.append({'arch': arch, 'target': tgt,
                                 'nRMSE': np.nan, 'R2': np.nan,
                                 'AnomalyRate': np.nan, 'SeqAcc': np.nan,
                                 'Composite': np.nan})
            else:
                records.append({'arch': arch, 'target': tgt,
                                 'nRMSE':       res['nrmse'],
                                 'R2':          res['r2'],
                                 'AnomalyRate': res['anomaly']['anomaly_rate'],
                                 'SeqAcc':      res['seq_acc']['seq_score'],
                                 'Composite':   composite(res)})
    summary = pd.DataFrame(records)

    fig = plt.figure(figsize=(28, 24))
    fig.patch.set_facecolor('white')
    gs  = gridspec.GridSpec(3, 2, figure=fig, hspace=0.45, wspace=0.35)
    ax_h  = fig.add_subplot(gs[0, :])
    ax_n  = fig.add_subplot(gs[1, 0])
    ax_r  = fig.add_subplot(gs[1, 1])
    ax_a  = fig.add_subplot(gs[2, 0])
    ax_s  = fig.add_subplot(gs[2, 1])
    fig.suptitle('Model Selection Dashboard  |  RNN vs LSTM vs GRU',
                 fontsize=20, fontweight='bold', y=1.01)

    pv = summary.pivot(index='arch', columns='target', values='Composite')
    sns.heatmap(pv, ax=ax_h, annot=True, fmt='.1f', cmap='RdYlGn_r',
                linewidths=0.6, linecolor='white',
                cbar_kws={'label': 'Composite Score (lower = better)'})
    ax_h.set_title(
        'Composite Heatmap  '
        '(nRMSE×0.35 + R²pen×0.25 + Anomaly×0.25 + SeqPen×0.15)',
        fontweight='bold', fontsize=11)
    ax_h.set_xlabel(''); ax_h.set_ylabel('Architecture', fontweight='bold')
    ax_h.tick_params(axis='x', rotation=30)

    x    = np.arange(len(tgt_list))
    w    = 0.25
    offs = [-w, 0, w]
    xlabels = [t.replace(' (%)','\n(%)').replace('Efficiency ','Eff.')
               for t in tgt_list]

    for ax, metric, ylabel in [
        (ax_n, 'nRMSE',       'nRMSE (%)'),
        (ax_r, 'R2',          'R² Score'),
        (ax_a, 'AnomalyRate', 'Anomaly Rate (%)'),
        (ax_s, 'SeqAcc',      'Seq. Acc. (%)'),
    ]:
        for i, arch in enumerate(arch_list):
            vals = [summary[(summary.arch == arch) &
                            (summary.target == t)][metric].values[0]
                    for t in tgt_list]
            vals = np.array(vals, dtype=float)
            bars = ax.bar(x+offs[i], vals, w, label=arch,
                          color=MODEL_CONFIGS[arch]['color'], alpha=0.85)
            for b in bars:
                h = b.get_height()
                if np.isfinite(h) and h > 0:
                    ax.text(b.get_x()+b.get_width()/2, h+0.3,
                            f'{h:.1f}', ha='center', fontsize=6.5)
        ax.set_xticks(x); ax.set_xticklabels(xlabels, fontsize=8)
        ax.set_ylabel(ylabel, fontsize=9)
        ax.set_title(ylabel, fontweight='bold')
        ax.legend(fontsize=8); ax.grid(axis='y', alpha=0.35)

    plt.savefig('/content/model_selection_dashboard.png', dpi=150, bbox_inches='tight')
    plt.show()
    print("✅ Saved: model_selection_dashboard.png")
    return summary


# ── 9J: SHAP + Metrics joint publication figure ───────────────────────────────
def plot_shap_metrics_joint(rba, summary):
    arch_list = list(MODEL_CONFIGS.keys())
    tgt_list  = TARGETS

    fig = plt.figure(figsize=(30, 15))
    fig.patch.set_facecolor('white')
    gs  = gridspec.GridSpec(2, 4, figure=fig, hspace=0.55, wspace=0.42)
    fig.suptitle('SHAP + Metrics Joint Summary Dashboard',
                 fontsize=18, fontweight='bold', y=1.02)

    metric_cfg = [
        ('nRMSE',       'nRMSE (%)',       'YlOrRd'),
        ('R2',          'R² Score',         'YlGn'),
        ('AnomalyRate', 'Anomaly Rate (%)', 'OrRd'),
        ('SeqAcc',      'Seq. Acc. (%)',    'BuGn'),
    ]

    # Row 1: metric heatmaps
    for col, (metric, title, cmap) in enumerate(metric_cfg):
        ax = fig.add_subplot(gs[0, col])
        pv = summary.pivot(index='arch', columns='target', values=metric)
        sns.heatmap(pv, ax=ax, annot=True, fmt='.1f', cmap=cmap,
                    linewidths=0.5, linecolor='white', cbar=False)
        ax.set_title(title, fontweight='bold', fontsize=11)
        ax.set_xlabel(''); ax.set_ylabel('Arch', fontsize=9)
        ax.tick_params(axis='x', rotation=35, labelsize=8)

    # Row 2: SHAP overall mean |value| per arch (all targets pooled)
    for col, arch in enumerate(arch_list):
        ax = fig.add_subplot(gs[1, col])
        combined = {}
        for tgt in tgt_list:
            res = rba[arch].get(tgt)
            if res is None or res.get('shap_2d') is None: continue
            mv = res['shap_2d'].mean(axis=0)
            for i, nm in enumerate(res['feature_names']):
                combined.setdefault(nm, []).append(mv[i])

        if not combined:
            ax.text(0.5, 0.5, 'SHAP N/A', ha='center', transform=ax.transAxes)
            continue

        feat_mean = {k: np.mean(v) for k, v in combined.items()}
        top_f     = sorted(feat_mean, key=feat_mean.get, reverse=True)[:10]
        vals      = [feat_mean[f] for f in top_f]
        pal       = plt.cm.Blues(np.linspace(0.35, 0.9, len(top_f)))
        ax.barh(top_f[::-1], vals[::-1], color=pal, alpha=0.9, edgecolor='white')
        ax.set_title(f'SHAP Overall\n{arch}', fontweight='bold',
                     color=MODEL_CONFIGS[arch]['color'], fontsize=11)
        ax.set_xlabel('mean |SHAP| (all targets)')
        ax.grid(axis='x', alpha=0.35)

    plt.savefig('/content/SHAP_metrics_joint.png', dpi=150, bbox_inches='tight')
    plt.show()
    print("✅ Saved: SHAP_metrics_joint.png")


# ─────────────────────────────────────────────────────────────────────────────
# 10.  WINNER TABLE
# ─────────────────────────────────────────────────────────────────────────────
def print_winner_table(summary):
    print("\n" + "=" * 92)
    print("  MODEL SELECTION — BEST ARCHITECTURE PER TARGET  (Composite score)")
    print("=" * 92)
    cw = [28, 6, 8, 6, 13, 10, 14]
    print(f"{'Target':<{cw[0]}} {'Arch':<{cw[1]}} {'nRMSE%':<{cw[2]}} "
          f"{'R²':<{cw[3]}} {'AnomalyRate%':<{cw[4]}} "
          f"{'SeqAcc%':<{cw[5]}} {'Composite':<{cw[6]}}")
    print("-" * 92)

    votes = {a: 0 for a in MODEL_CONFIGS}
    for tgt in TARGETS:
        sub = summary[summary.target == tgt].dropna(subset=['Composite'])
        if sub.empty: continue
        best = sub.loc[sub['Composite'].idxmin()]
        votes[best['arch']] += 1
        print(f"{tgt:<{cw[0]}} {best['arch']:<{cw[1]}} "
              f"{best['nRMSE']:<{cw[2]}.2f} {best['R2']:<{cw[3]}.3f} "
              f"{best['AnomalyRate']:<{cw[4]}.1f} "
              f"{best['SeqAcc']:<{cw[5]}.1f} "
              f"{best['Composite']:<{cw[6]}.2f}")

    best_overall = max(votes, key=votes.get)
    print("=" * 92)
    print(f"\n  OVERALL BEST → {best_overall}   |   Votes: {votes}")
    print("\n  Composite formula (lower = better):")
    print("    0.35 × nRMSE  +  0.25 × (1−R²)×100  "
          "+  0.25 × AnomalyRate  +  0.15 × (100−SeqAcc)")
    print("=" * 92)

    pd.set_option('display.float_format', '{:.3f}'.format)
    pd.set_option('display.max_columns', 20)
    pd.set_option('display.width', 130)
    print("\n  FULL TABLE:")
    print(summary.sort_values(['target', 'Composite']).to_string(index=False))


# ─────────────────────────────────────────────────────────────────────────────
# 11.  MAIN
# ─────────────────────────────────────────────────────────────────────────────
def main():
    # ── Step 1: load + engineer + clean ──────────────────────────────────────
    print("\n📁 Step 1/6 — Data loading & preprocessing")
    df = load_hybrid_data()
    if len(df) < 200:
        print("❌ Insufficient data!"); return
    df = engineer_features(df,
                           system_capacity_wp=TRAIN_CFG['system_capacity_wp'],
                           panel_area_m2=TRAIN_CFG['panel_area_m2'])
    df = clean_data(df)

    # ── Step 2: train 18 models ───────────────────────────────────────────────
    print("\n🤖 Step 2/6 — Training 3 arch × 6 targets = 18 models")
    print("─" * 72)

    rba = {arch: {} for arch in MODEL_CONFIGS}
    for arch in MODEL_CONFIGS:
        cfg = MODEL_CONFIGS[arch]
        print(f"\n🔷 {arch}   seq={cfg['seq']}   units={cfg['units']}")
        for tgt in TARGETS:
            if tgt in df.columns:
                rba[arch][tgt] = train_model(df, tgt, arch)
            else:
                print(f"   ⚠️  Column '{tgt}' not found — skipping")
                rba[arch][tgt] = None

    # ── Step 3: prediction plots ──────────────────────────────────────────────
    print("\n📈 Step 3/6 — Prediction time-series plots")
    plot_predictions(rba)

    # ── Step 4: permutation XAI ───────────────────────────────────────────────
    print("\n🔍 Step 4/6 — Permutation Feature Importance")
    plot_perm_xai(rba, top_n=10)

    # ── Step 5: SHAP ──────────────────────────────────────────────────────────
    print("\n🎯 Step 5/6 — SHAP XAI (bar + beeswarm + temporal + cross-model)")
    plot_shap_bar(rba)
    plot_shap_beeswarm(rba)
    plot_shap_temporal(rba)
    plot_shap_cross_model(rba)

    # ── Step 5b: anomaly + seq accuracy ───────────────────────────────────────
    print("\n🔴 Step 5b — Anomaly Dashboard + Sequence Accuracy")
    plot_anomaly_dashboard(rba)
    plot_seq_accuracy(rba)

    # ── Step 6: model selection ────────────────────────────────────────────────
    print("\n🏆 Step 6/6 — Model Selection Dashboard + Winner Table")
    summary = plot_model_selection_dashboard(rba)
    plot_shap_metrics_joint(rba, summary)
    print_winner_table(summary)

    summary.to_csv('/content/model_comparison_results.csv', index=False)
    print("\n💾 Saved → /content/model_comparison_results.csv")
    print("\n" + "=" * 80)
    print("✅  ALL DONE!")
    print("=" * 80)


if __name__ == '__main__':
    main()
```
