```python
import os
os.environ['CUDA_VISIBLE_DEVICES'] = '-1'
os.environ['TF_CPP_MIN_LOG_LEVEL'] = '3'

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.dates as mdates
import seaborn as sns
from itertools import product as iterproduct
from sklearn.preprocessing import MinMaxScaler
from sklearn.metrics import mean_squared_error, r2_score
import warnings
warnings.filterwarnings('ignore')

# SET WHITE THEME DI AWAL
plt.style.use('default')
plt.rcParams.update({
    'figure.facecolor': 'white',
    'axes.facecolor': 'white',
    'axes.edgecolor': 'black',
    'axes.labelcolor': 'black',
    'axes.titlecolor': 'black',
    'xtick.color': 'black',
    'ytick.color': 'black',
    'text.color': 'black',
    'legend.facecolor': 'white',
    'legend.edgecolor': 'black',
    'legend.labelcolor': 'black',
    'grid.color': '#f0f0f0',
    'grid.alpha': 0.3
})

print("✅ HYBRID PLTS — LSTM LOADED!")

# ── CHANGE 1: Import LSTM instead of SimpleRNN ──────────────────────────────
try:
    from tensorflow.keras.models import Sequential
    from tensorflow.keras.layers import LSTM, Dense, Dropout           # ← LSTM
    from tensorflow.keras.optimizers import Adam
    from tensorflow.keras.callbacks import EarlyStopping
    TF_AVAILABLE = True
    print("✅ TensorFlow OK")
except Exception as e:
    TF_AVAILABLE = False
    print(f"⚠️ TensorFlow not available: {e}")
    Sequential = LSTM = Dense = Dropout = Adam = EarlyStopping = None  # ← LSTM

# =============================================================================
# EXPERIMENT GRID — LSTM
# Model Type | Layers    | Seq Length  | Units per Layer
# LSTM       | 1, 2, 3, 4 | 32, 64, 128 | 64, 128, 256
# Total kombinasi: 4 × 3 × 3 = 36 eksperimen per target
# =============================================================================
_LAYERS_LIST   = [1, 2, 3, 4]
_SEQLEN_LIST   = [32, 64, 128]
_UNITS_LIST    = [64, 128, 256]

EXPERIMENT_GRID = list(iterproduct(_LAYERS_LIST, _SEQLEN_LIST, _UNITS_LIST))
# Setiap elemen: (n_layers, seq_len, units)

BEST_CONFIG = {
    'merge_tolerance'    : '10m',
    'targets'            : ['Power PV', 'PR (%)'],
    'n_experiments'      : len(EXPERIMENT_GRID),   # 36
    'system_capacity_wp' : 2250,
    'panel_area_m2'      : 10.893,
    'plot_predictions'   : True,
    'plot_start'         : '2026-03-27',
    'plot_end'           : '2026-03-31 23:59:59',
}

# =============================================================================
# LEAKAGE MAP
# =============================================================================
LEAKAGE_MAP = {
    'Power PV': [
        'Power PV',
        'Efficiency PV (%)',
        'Efficiency System (%)',
        'PR (%)',
        'P_pv',
    ],
    'PR (%)': [],
}

# =============================================================================
# XAI EXCLUDE MAP
# =============================================================================
XAI_EXCLUDE = {
    'Power PV': [],
    'PR (%)'  : ['PR (%)', 'PR_smooth'],
}

# =============================================================================
# 1. DATA LOADING & MERGING
# =============================================================================
def load_hybrid_data(
        env_file='/content/Lingkungan.csv',
        dc_file ='/content/Elektrik DC.csv',
        ac_file ='/content/Elektrik AC.csv'):
    print("📁 Loading 3 files...")
    df_env = pd.read_csv(env_file)
    df_dc  = pd.read_csv(dc_file)
    df_ac  = pd.read_csv(ac_file)

    def parse_datetime(df):
        df = df.copy()
        df['datetime'] = pd.to_datetime(
            df['Date'].astype(str) + ' ' + df['Time'].astype(str), errors='coerce')
        return df.dropna(subset=['datetime']).sort_values('datetime')

    df_env = parse_datetime(df_env)
    df_dc  = parse_datetime(df_dc)
    df_ac  = parse_datetime(df_ac)
    for frame in (df_env, df_dc, df_ac):
        frame.set_index('datetime', inplace=True)

    tol = pd.Timedelta(BEST_CONFIG['merge_tolerance'])
    df = pd.merge_asof(df_dc.sort_index(), df_env.sort_index(),
                       left_index=True, right_index=True,
                       tolerance=tol, direction='nearest')
    df = pd.merge_asof(df.sort_index(), df_ac.sort_index(),
                       left_index=True, right_index=True,
                       tolerance=tol, direction='nearest')

    df['hour']       = df.index.hour
    df['minute']     = df.index.minute
    df['dayofweek']  = df.index.dayofweek
    df['month']      = df.index.month
    df['is_daytime'] = ((df.index.hour >= 6) & (df.index.hour <= 18)).astype(int)
    df['hour_sin']   = np.sin(2 * np.pi * df.index.hour   / 24)
    df['hour_cos']   = np.cos(2 * np.pi * df.index.hour   / 24)
    df['minute_sin'] = np.sin(2 * np.pi * df.index.minute / 60)
    df['minute_cos'] = np.cos(2 * np.pi * df.index.minute / 60)

    print(f"✅ Merged: {len(df)} rows")
    return df

# =============================================================================
# 2. FEATURE ENGINEERING
# =============================================================================
def engineer_hybrid_features(df,
                              system_capacity_wp=BEST_CONFIG['system_capacity_wp'],
                              panel_area_m2=BEST_CONFIG['panel_area_m2']):
    df = df.copy()
    for col in df.columns:
        df[col] = pd.to_numeric(df[col], errors='coerce')

    df['delta_t'] = df.index.to_series().diff().dt.total_seconds() / 3600

    df['P_ac'] = df['V AC'] * df['I AC']
    df['E_ac'] = df['P_ac'] * df['delta_t']

    if 'Irradiance' in df.columns and 'Power PV' in df.columns:
        df['Efficiency PV (%)'] = (
            df['Power PV'] / (df['Irradiance'] * panel_area_m2)) * 100

    df['is_charging'] = (
        (df.index.hour >= 6) & (df.index.hour < 18) &
        (df['V25 (V)'].fillna(0) > 0) &
        (df['I ACS758 (A)'].fillna(0) > 0)
    )
    df['P_batt'] = df['V25 (V)'] * df['I ACS758 (A)']
    df['E_batt'] = df['P_batt'] * df['delta_t']

    df['E_charge']    = np.nan
    df['E_discharge'] = np.nan
    df.loc[ df['is_charging'], 'E_charge']    = df.loc[ df['is_charging'], 'E_batt']
    df.loc[~df['is_charging'], 'E_discharge'] = df.loc[~df['is_charging'], 'E_batt']

    df['E_charge_total']    = df['E_charge'].fillna(0).rolling('24h', min_periods=1).sum().shift(1)
    df['E_discharge_total'] = df['E_discharge'].fillna(0).rolling('24h', min_periods=1).sum().shift(1)
    df['Efficiency Battery (%)'] = (
        df['E_discharge_total'] / df['E_charge_total'].replace(0, np.nan)) * 100

    df['P_pv']             = df['V220 (V)'] * df['I ACS712 (A)']
    df['P_batt_discharge'] = (df['V25 (V)'] * df['I ACS758 (A)']).clip(lower=0)
    df['E_dc_total']       = df['P_pv'] * df['delta_t']

    df['E_dc_total_24h'] = df['E_dc_total'].rolling('24h', min_periods=1).sum().shift(1)
    df['E_ac_24h']       = df['E_ac'].rolling('24h', min_periods=1).sum().shift(1)

    df['Efficiency Inv (%)'] = (df['E_ac_24h'] / df['E_dc_total_24h']) * 100
    df['Efficiency Inv (%)'] = df['Efficiency Inv (%)'].replace([np.inf, -np.inf], np.nan)

    df['Efficiency Temp (%)']   = (1 - 0.004 * (df['Temp PV (C)'].fillna(25) - 25)) * 100
    df['Efficiency System (%)'] = (df['P_ac'] / (df['Irradiance'] * panel_area_m2)) * 100

    df['Irrad_kWh']       = (df['Irradiance'] * df['delta_t']) / 1000
    df['Irrad_total_24h'] = df['Irrad_kWh'].rolling('24h', min_periods=1).sum().shift(1)
    df['PR (%)']          = (
        df['E_ac_24h'] / (system_capacity_wp * df['Irrad_total_24h'])) * 100

    df['PR_smooth'] = df['PR (%)'].rolling(3, min_periods=1).mean()

    eff_cols = [
        'Efficiency PV (%)', 'Efficiency Battery (%)', 'Efficiency Inv (%)',
        'Efficiency Temp (%)', 'Efficiency System (%)', 'PR (%)', 'PR_smooth'
    ]
    for col in eff_cols:
        if col in df.columns:
            df[col] = df[col].fillna(df[col].median())

    df.loc[df['Irradiance'] < 50, 'Efficiency PV (%)']  = np.nan
    df.loc[df['P_pv'] < 10,       'Efficiency Inv (%)'] = np.nan

    df['Efficiency PV (%)']      = df['Efficiency PV (%)'].clip(0, 25)
    df['Efficiency Battery (%)'] = df['Efficiency Battery (%)'].clip(0, 100)
    df['Efficiency Inv (%)']     = df['Efficiency Inv (%)'].clip(0, 100)
    df['Efficiency Temp (%)']    = df['Efficiency Temp (%)'].clip(50, 100)
    df['Efficiency System (%)']  = df['Efficiency System (%)'].clip(0, 100)
    df['PR (%)']                 = df['PR (%)'].clip(0, 100)
    df['PR_smooth']              = df['PR_smooth'].clip(0, 100)

    print("✅ All efficiencies + PR_smooth calculated!")
    return df

# =============================================================================
# 3. DATA CLEANING
# =============================================================================
def clean_data(df):
    df = df.copy()
    sensor_cols = ['V25 (V)', 'I ACS758 (A)', 'V AC', 'I AC',
                   'Irradiance', 'Power PV', 'Temp PV (C)']
    for col in sensor_cols:
        if col in df.columns:
            df[col] = pd.to_numeric(df[col], errors='coerce')

    numeric_cols = df.select_dtypes(include=[np.number]).columns
    for col in numeric_cols:
        if df[col].isnull().any():
            df[col] = df[col].fillna(df[col].median())
        df[col] = np.nan_to_num(df[col],
                                posinf=np.finfo(np.float64).max,
                                neginf=np.finfo(np.float64).min)
        q1, q3  = df[col].quantile([0.25, 0.75])
        iqr     = q3 - q1
        df[col] = df[col].clip(q1 - 1.5 * iqr, q3 + 1.5 * iqr)
    return df

# =============================================================================
# 4. XAI
# =============================================================================
def compute_xai(model, X_test_scaled, feature_names, df, target_col):
    print(f"\n🧠 XAI for '{target_col}'...")

    base_exclude  = {target_col}
    extra_exclude = set(XAI_EXCLUDE.get(target_col, []))
    all_exclude   = base_exclude | extra_exclude

    valid_idx      = [i for i, f in enumerate(feature_names) if f not in all_exclude]
    valid_features = [feature_names[i] for i in valid_idx]

    excluded_found = [f for f in feature_names if f in all_exclude]
    if excluded_found:
        print(f"   🚫 Excluded from XAI display: {excluded_found}")
    print(f"   ✅ XAI features remaining: {len(valid_features)}")

    n_samples = min(20, len(X_test_scaled))
    baseline  = model.predict(X_test_scaled[:n_samples], verbose=0).mean()

    ablation = {}
    for i, feat in zip(valid_idx, valid_features):
        X_tmp = X_test_scaled[:n_samples].copy()
        fill  = X_test_scaled[:n_samples, :, i].mean()
        X_tmp[:, :, i] = 0.0 if np.isnan(fill) else fill
        ablation[feat] = abs(baseline - model.predict(X_tmp, verbose=0).mean())

    target_s = df[target_col].dropna()
    corr_imp = {}
    for feat in valid_features:
        if feat not in df.columns:
            corr_imp[feat] = 0.0; continue
        fs = df[feat].dropna()
        common = target_s.index.intersection(fs.index)
        if len(common) > 10 and fs.std() != 0 and target_s.std() != 0:
            c = target_s.loc[common].corr(fs.loc[common])
            corr_imp[feat] = 0.0 if np.isnan(c) else abs(c)
        else:
            corr_imp[feat] = 0.0

    combined = {f: ablation.get(f, 0) * 0.7 + corr_imp.get(f, 0) * 0.3
                for f in valid_features}
    combined = {k: v for k, v in combined.items() if np.isfinite(v)}
    if not combined:
        return pd.DataFrame({'Feature': [], 'Impact(%)': []})

    total  = sum(combined.values()) or 1
    xai_df = pd.DataFrame({
        'Feature'  : list(combined.keys()),
        'Impact(%)': [v / total * 100 for v in combined.values()]
    }).sort_values('Impact(%)', ascending=False).head(10).fillna(0)
    print(xai_df.to_string(index=False))
    return xai_df


def plot_xai(xai_df, target_name):
    plt.figure(figsize=(10, 6))
    sns.barplot(x='Impact(%)', y='Feature', data=xai_df, palette='viridis')
    plt.title(f'Feature Importance — {target_name} (XAI)', fontweight='bold')
    plt.xlabel('Impact (%)'); plt.ylabel('Feature')
    plt.grid(axis='x', alpha=0.3); plt.tight_layout(); plt.show()


def plot_xai_scatter_features(xai_df, target_name, res, df):
    ts        = pd.DatetimeIndex(res['test_timestamps'])
    y_true    = res['y_test_orig']
    resid     = y_true - res['y_pred']
    n_top     = min(5, len(xai_df))
    top_feats = xai_df['Feature'].tolist()[:n_top]

    fig, axes = plt.subplots(1, n_top, figsize=(5 * n_top, 5))
    fig.patch.set_facecolor('white')
    if n_top == 1:
        axes = [axes]
    fig.suptitle(f'🔍 XAI — Top Feature Scatter vs Target: {target_name}',
                 fontsize=14, fontweight='bold', color='black', y=1.01)

    for ax, feat in zip(axes, top_feats):
        ax.set_facecolor('white')
        for sp in ax.spines.values():
            sp.set_edgecolor('black')
        if feat not in df.columns:
            ax.text(0.5, 0.5, f'{feat}\n(not in df)', ha='center',
                    va='center', transform=ax.transAxes, color='red')
            continue
        try:
            feat_vals = df[feat].reindex(ts).values
        except Exception:
            feat_vals = np.full(len(ts), np.nan)

        valid = ~np.isnan(feat_vals)
        if valid.sum() < 5:
            ax.text(0.5, 0.5, f'{feat}\n(insufficient data)', ha='center',
                    va='center', transform=ax.transAxes, color='red')
            continue

        sc = ax.scatter(feat_vals[valid], y_true[valid],
                        c=np.abs(resid[valid]), cmap='RdYlGn_r',
                        alpha=0.6, s=18, edgecolors='none')
        try:
            z = np.polyfit(feat_vals[valid], y_true[valid], 1)
            p = np.poly1d(z)
            x_line = np.linspace(feat_vals[valid].min(), feat_vals[valid].max(), 100)
            ax.plot(x_line, p(x_line), 'k--', lw=1.5, alpha=0.7, label='Trend')
        except Exception:
            pass

        cb = fig.colorbar(sc, ax=ax, pad=0.02, shrink=0.85)
        cb.set_label('|Residual|', color='black', fontsize=8)
        plt.setp(cb.ax.yaxis.get_ticklabels(), color='black', fontsize=7)

        impact_val = xai_df.loc[xai_df['Feature'] == feat, 'Impact(%)'].values
        impact_str = f"  ({impact_val[0]:.1f}%)" if len(impact_val) else ""
        ax.set_xlabel(feat + impact_str, color='black', fontsize=8)
        ax.set_ylabel(target_name, color='black', fontsize=9)
        ax.set_title(feat, color='black', fontsize=9, fontweight='bold')
        ax.tick_params(colors='black', labelsize=7)
        ax.grid(True, alpha=0.2, color='#e0e0e0')

    plt.tight_layout()
    plt.show()


def plot_xai_residual_by_quartile(xai_df, target_name, res, df):
    ts      = pd.DatetimeIndex(res['test_timestamps'])
    resid   = res['y_test_orig'] - res['y_pred']
    n_top   = min(3, len(xai_df))
    top3    = xai_df['Feature'].tolist()[:n_top]

    fig, axes = plt.subplots(1, n_top, figsize=(7 * n_top, 6))
    fig.patch.set_facecolor('white')
    if n_top == 1:
        axes = [axes]
    fig.suptitle(f'📦 XAI — Residual per Feature Quartile: {target_name}\n'
                 f'(Positif = under-predict, Negatif = over-predict)',
                 fontsize=13, fontweight='bold', color='black')

    q_labels = ['Q1\n(Rendah)', 'Q2', 'Q3', 'Q4\n(Tinggi)']
    for ax, feat in zip(axes, top3):
        ax.set_facecolor('white')
        for sp in ax.spines.values():
            sp.set_edgecolor('black')
        if feat not in df.columns:
            ax.text(0.5, 0.5, f'{feat}\n(not in df)', ha='center',
                    va='center', transform=ax.transAxes, color='red')
            continue
        try:
            feat_vals = df[feat].reindex(ts).values
        except Exception:
            feat_vals = np.full(len(ts), np.nan)

        valid = ~np.isnan(feat_vals)
        if valid.sum() < 20:
            ax.text(0.5, 0.5, f'{feat}\n(data terlalu sedikit)', ha='center',
                    va='center', transform=ax.transAxes, color='red')
            continue
        try:
            q_cut = pd.qcut(feat_vals[valid], q=4, labels=q_labels, duplicates='drop')
        except Exception:
            ax.text(0.5, 0.5, f'{feat}\n(variance rendah)', ha='center',
                    va='center', transform=ax.transAxes, color='gray')
            continue

        qdf = pd.DataFrame({'Kuartil': q_cut, 'Residual': resid[valid]})
        sns.boxplot(x='Kuartil', y='Residual', data=qdf,
                    palette='coolwarm', ax=ax,
                    flierprops=dict(marker='o', markersize=3, alpha=0.5))
        ax.axhline(0, color='red', lw=1.8, ls='--', alpha=0.8, label='Zero line')

        for i, ql in enumerate(q_cut.categories):
            med = qdf.loc[qdf['Kuartil'] == ql, 'Residual'].median()
            if not np.isnan(med):
                ax.text(i, med, f'{med:.2f}', ha='center', va='bottom',
                        fontsize=8, color='black', fontweight='bold')

        impact_val = xai_df.loc[xai_df['Feature'] == feat, 'Impact(%)'].values
        impact_str = f" [{impact_val[0]:.1f}%]" if len(impact_val) else ""
        ax.set_title(f'{feat}{impact_str}', color='black', fontweight='bold', fontsize=10)
        ax.set_xlabel('Kuartil Nilai Fitur', color='black')
        ax.set_ylabel('Residual (Actual − Prediction)', color='black')
        ax.tick_params(colors='black')
        ax.legend(fontsize=8, facecolor='white', edgecolor='black')
        ax.grid(axis='y', alpha=0.25, color='#e0e0e0')

    plt.tight_layout()
    plt.show()


def plot_xai_temporal_error(target_name, res):
    ts    = pd.DatetimeIndex(res['test_timestamps'])
    resid = res['y_test_orig'] - res['y_pred']

    df_err = pd.DataFrame({
        'Hour'     : ts.hour,
        'DayOfWeek': ts.dayofweek,
        'AbsError' : np.abs(resid),
        'Residual' : resid,
    })

    day_names = ['Sen', 'Sel', 'Rab', 'Kam', 'Jum', 'Sab', 'Min']

    fig, axes = plt.subplots(1, 3, figsize=(22, 6))
    fig.patch.set_facecolor('white')
    fig.suptitle(f'⏰ XAI — Temporal Error Pattern: {target_name}\n'
                 f'(Deteksi jam & hari yang paling sering error)',
                 fontsize=13, fontweight='bold', color='black')

    ax1 = axes[0]
    pivot = (df_err.groupby(['Hour', 'DayOfWeek'])['AbsError']
             .mean()
             .unstack(fill_value=np.nan))
    pivot.columns = [day_names[d] for d in pivot.columns if d < len(day_names)]
    sns.heatmap(pivot, cmap='YlOrRd', ax=ax1,
                linewidths=0.3,
                cbar_kws={'label': 'Mean |Error|', 'shrink': 0.85},
                annot=(pivot.shape[0] <= 24 and pivot.shape[1] <= 7))
    ax1.set_title('Mean |Error| per Jam & Hari', color='black', fontweight='bold')
    ax1.set_xlabel('Hari dalam Seminggu', color='black')
    ax1.set_ylabel('Jam dalam Sehari', color='black')
    ax1.tick_params(colors='black', labelsize=8)
    for sp in ax1.spines.values():
        sp.set_edgecolor('black')

    ax2 = axes[1]
    ax2.set_facecolor('white')
    hourly_mean = df_err.groupby('Hour')['AbsError'].mean()
    hourly_std  = df_err.groupby('Hour')['AbsError'].std().fillna(0)
    ax2.bar(hourly_mean.index, hourly_mean.values,
            color='#4fc3f7', edgecolor='black', alpha=0.85, zorder=3, label='Mean |Error|')
    ax2.errorbar(hourly_mean.index, hourly_mean.values,
                 yerr=hourly_std.values, fmt='none',
                 color='#d32f2f', lw=1.5, capsize=3, label='±1 Std Dev')
    peak_h = hourly_mean.idxmax()
    ax2.axvline(peak_h, color='red', lw=1.5, ls='--', alpha=0.7,
                label=f'Peak error: {peak_h:02d}:00')
    ax2.set_title('Mean |Error| per Jam', color='black', fontweight='bold')
    ax2.set_xlabel('Jam', color='black')
    ax2.set_ylabel('Mean |Error|', color='black')
    ax2.tick_params(colors='black')
    ax2.legend(fontsize=8, facecolor='white', edgecolor='black')
    ax2.grid(axis='y', alpha=0.3, color='#e0e0e0', zorder=0)
    for sp in ax2.spines.values():
        sp.set_edgecolor('black')

    ax3 = axes[2]
    ax3.set_facecolor('white')
    hourly_res  = df_err.groupby('Hour')['Residual'].mean()
    colors_bar  = ['#69f0ae' if v >= 0 else '#ff7043' for v in hourly_res.values]
    ax3.bar(hourly_res.index, hourly_res.values, color=colors_bar,
            edgecolor='black', alpha=0.85, zorder=3)
    ax3.axhline(0, color='black', lw=1.5, alpha=0.8)
    ax3.fill_between(hourly_res.index, hourly_res.values, 0,
                     where=(hourly_res.values >= 0), alpha=0.2, color='#69f0ae',
                     label='Under-predict (positif)')
    ax3.fill_between(hourly_res.index, hourly_res.values, 0,
                     where=(hourly_res.values < 0),  alpha=0.2, color='#ff7043',
                     label='Over-predict (negatif)')
    ax3.set_title('Mean Signed Residual per Jam\n(Arah bias prediksi)',
                  color='black', fontweight='bold')
    ax3.set_xlabel('Jam', color='black')
    ax3.set_ylabel('Mean Residual (Actual − Pred)', color='black')
    ax3.tick_params(colors='black')
    ax3.legend(fontsize=8, facecolor='white', edgecolor='black')
    ax3.grid(axis='y', alpha=0.3, color='#e0e0e0', zorder=0)
    for sp in ax3.spines.values():
        sp.set_edgecolor('black')

    plt.tight_layout()
    plt.show()


def plot_xai_comparison(all_xai_results, target_name):
    n_models = len(all_xai_results)
    if n_models < 2:
        print("⚠️ Perlu ≥2 model untuk perbandingan XAI")
        return

    sorted_models = sorted(
        all_xai_results.items(),
        key=lambda x: x[1]['r2'],
        reverse=True
    )[:5]
    n_show = len(sorted_models)

    fig, axes = plt.subplots(1, n_show, figsize=(6 * n_show, 8))
    fig.patch.set_facecolor('white')
    if n_show == 1:
        axes = [axes]

    fig.suptitle(f'🔍 XAI Comparison — {target_name}\n'
                 f'(Mengapa konfigurasi tertentu lebih baik?)',
                 fontsize=14, fontweight='bold', color='black')

    all_features = set()
    for _, data in sorted_models:
        all_features.update(data['xai_df']['Feature'].tolist())
    all_features = list(all_features)[:15]

    for idx, (model_name, data) in enumerate(sorted_models):
        ax = axes[idx]
        ax.set_facecolor('white')
        for sp in ax.spines.values():
            sp.set_edgecolor('black')

        xai_filtered = data['xai_df'][data['xai_df']['Feature'].isin(all_features)]

        if xai_filtered.empty:
            ax.text(0.5, 0.5, 'No Data', ha='center', va='center',
                    transform=ax.transAxes, color='red')
            continue

        sns.barplot(x='Impact(%)', y='Feature', data=xai_filtered,
                    palette='viridis', ax=ax)

        ax.set_title(f'{model_name}\nR²={data["r2"]:.3f} | RMSE={data["rmse"]:.3f}',
                     color='black', fontweight='bold', fontsize=9)
        ax.set_xlabel('Impact (%)', color='black', fontsize=8)

        if idx != 0:
            ax.set_yticks([])
            ax.set_ylabel('')
        else:
            ax.set_ylabel('Feature', color='black')

        ax.tick_params(colors='black', labelsize=7)
        ax.grid(axis='x', alpha=0.3, color='#e0e0e0')

    plt.tight_layout()
    plt.show()


def plot_config_impact_analysis(summary_df, target_name):
    target_df = summary_df[summary_df['Target'] == target_name]
    if target_df.empty:
        print(f"⚠️ Tidak ada data untuk target '{target_name}'")
        return

    fig, axes = plt.subplots(1, 3, figsize=(18, 6))
    fig.patch.set_facecolor('white')
    for ax in axes:
        ax.set_facecolor('white')
        for sp in ax.spines.values():
            sp.set_edgecolor('black')

    ax1 = axes[0]
    layers_summary = (target_df.groupby('Layers')['R2']
                      .agg(['mean', 'std'])
                      .reset_index()
                      .rename(columns={'mean': 'R2_mean', 'std': 'R2_std'}))
    layers_summary['R2_std'] = layers_summary['R2_std'].fillna(0)
    ax1.errorbar(layers_summary['Layers'], layers_summary['R2_mean'],
                 yerr=layers_summary['R2_std'], fmt='o-', capsize=5,
                 color='#4fc3f7', lw=2, markersize=8)
    ax1.set_xlabel('Jumlah Layer', color='black')
    ax1.set_ylabel('R² (mean ± std)', color='black')
    ax1.set_title('Dampak Jumlah Layer terhadap Performa',
                  color='black', fontweight='bold')
    ax1.tick_params(colors='black')
    ax1.grid(True, alpha=0.3, color='#e0e0e0')

    ax2 = axes[1]
    seqlen_summary = (target_df.groupby('SeqLen')['R2']
                      .agg(['mean', 'std'])
                      .reset_index()
                      .rename(columns={'mean': 'R2_mean', 'std': 'R2_std'}))
    seqlen_summary['R2_std'] = seqlen_summary['R2_std'].fillna(0)
    ax2.errorbar(seqlen_summary['SeqLen'], seqlen_summary['R2_mean'],
                 yerr=seqlen_summary['R2_std'], fmt='s-', capsize=5,
                 color='#69f0ae', lw=2, markersize=8)
    ax2.set_xlabel('Sequence Length', color='black')
    ax2.set_ylabel('R² (mean ± std)', color='black')
    ax2.set_title('Dampak Sequence Length terhadap Performa',
                  color='black', fontweight='bold')
    ax2.tick_params(colors='black')
    ax2.grid(True, alpha=0.3, color='#e0e0e0')

    ax3 = axes[2]
    units_summary = (target_df.groupby('Units')['R2']
                     .agg(['mean', 'std'])
                     .reset_index()
                     .rename(columns={'mean': 'R2_mean', 'std': 'R2_std'}))
    units_summary['R2_std'] = units_summary['R2_std'].fillna(0)
    ax3.errorbar(units_summary['Units'], units_summary['R2_mean'],
                 yerr=units_summary['R2_std'], fmt='^-', capsize=5,
                 color='#ff7043', lw=2, markersize=8)
    ax3.set_xlabel('Jumlah Units', color='black')
    ax3.set_ylabel('R² (mean ± std)', color='black')
    ax3.set_title('Dampak Units terhadap Performa',
                  color='black', fontweight='bold')
    ax3.tick_params(colors='black')
    ax3.grid(True, alpha=0.3, color='#e0e0e0')

    fig.suptitle(f'📊 Hyperparameter Impact Analysis — {target_name}',
                 fontsize=14, fontweight='bold', color='black')
    plt.tight_layout()
    plt.show()


# =============================================================================
# 5. DATE-FILTERED VALIDATION PLOT
# =============================================================================
def plot_date_filtered(y_true, y_pred, timestamps, target_name,
                       rmse, nrmse, mape, r2,
                       n_layers, seq_len, units,
                       plot_start=None, plot_end=None):
    df_result = pd.DataFrame(
        {'Actual': y_true, 'Prediction': y_pred},
        index=pd.to_datetime(timestamps)
    )
    df_result['Residual'] = df_result['Actual'] - df_result['Prediction']

    if plot_start and plot_end:
        mask    = (df_result.index >= plot_start) & (df_result.index <= plot_end)
        plot_df = df_result.loc[mask] if mask.any() else df_result
        date_label = f"[{plot_start}  →  {plot_end[:10]}]"
    else:
        plot_df    = df_result
        date_label = "[Full Test Period]"

    if plot_df.empty:
        plot_df    = df_result
        date_label = "[Full Test Period — fallback]"

    fig, axes = plt.subplots(2, 2, figsize=(18, 11))
    fig.patch.set_facecolor('white')
    for ax in axes.flat:
        ax.set_facecolor('white')
        for sp in ax.spines.values():
            sp.set_edgecolor('black')

    # ── CHANGE 3: Update title label to LSTM ────────────────────────────────
    fig.suptitle(
        f'🎯 LSTM Prediction: {target_name}   {date_label}\n'
        f'Config L{n_layers}-S{seq_len}-U{units}   |   '
        f'RMSE={rmse:.3f}   nRMSE={nrmse:.1f}%   MAPE={mape:.1f}%   R²={r2:.3f}',
        fontsize=13, fontweight='bold', color='black', y=0.98
    )

    t = plot_df.index
    a = plot_df['Actual'].values
    p = plot_df['Prediction'].values
    r = plot_df['Residual'].values

    ax1 = axes[0, 0]
    ax1.plot(t, a, color='#4fc3f7', lw=2.5, label='Actual',     alpha=0.9)
    ax1.plot(t, p, color='#ff7043', lw=2.5, label='Prediction', alpha=0.9, ls='--')
    ax1.fill_between(t, a, p, alpha=0.18, color='#ffd54f', label='Error Band')
    ax1.set_title('Actual vs Prediction (filtered window)', color='black', fontweight='bold')
    ax1.tick_params(colors='black')
    ax1.xaxis.set_major_formatter(mdates.DateFormatter('%d-%b\n%H:%M'))
    ax1.legend(facecolor='white', labelcolor='black', fontsize=9, edgecolor='black')
    ax1.grid(True, alpha=0.2, color='#e0e0e0')

    ax2 = axes[0, 1]
    ax2.plot(t, r, color='#69f0ae', lw=2, alpha=0.8)
    ax2.axhline(0, color='black', lw=1.2, alpha=0.6)
    ax2.fill_between(t, r, 0, where=(r >= 0), alpha=0.25, color='#69f0ae')
    ax2.fill_between(t, r, 0, where=(r  < 0), alpha=0.25, color='#ff7043')
    ax2.set_title('Residuals (Actual − Prediction)', color='black', fontweight='bold')
    ax2.tick_params(colors='black')
    ax2.xaxis.set_major_formatter(mdates.DateFormatter('%d-%b\n%H:%M'))
    ax2.grid(True, alpha=0.2, color='#e0e0e0')

    ax3 = axes[1, 0]
    all_a = df_result['Actual'].values
    all_p = df_result['Prediction'].values
    sc = ax3.scatter(all_a, all_p, c=np.abs(df_result['Residual'].values),
                     cmap='plasma', alpha=0.5, s=18)
    lo, hi = min(all_a.min(), all_p.min()), max(all_a.max(), all_p.max())
    ax3.plot([lo, hi], [lo, hi], 'k--', lw=1.8, label='Perfect prediction', alpha=0.7)
    cb = fig.colorbar(sc, ax=ax3, pad=0.02)
    cb.set_label('|Residual|', color='black')
    plt.setp(cb.ax.yaxis.get_ticklabels(), color='black')
    ax3.set_xlabel('Actual', color='black'); ax3.set_ylabel('Predicted', color='black')
    ax3.set_title('Scatter: Full Test Set', color='black', fontweight='bold')
    ax3.tick_params(colors='black')
    ax3.legend(facecolor='white', labelcolor='black', fontsize=9, edgecolor='black')
    ax3.grid(True, alpha=0.2, color='#e0e0e0')

    ax4 = axes[1, 1]
    all_res = df_result['Residual'].values
    ax4.hist(all_res, bins=35, color='#ce93d8', edgecolor='black', alpha=0.8)
    ax4.axvline(all_res.mean(), color='#ff5252', lw=2.2, ls='--',
                label=f'Mean: {all_res.mean():.3f}')
    ax4.axvline(np.median(all_res), color='#ffeb3b', lw=2.2, ls=':',
                label=f'Median: {np.median(all_res):.3f}')
    ax4.set_title('Error Distribution — Full Test Set', color='black', fontweight='bold')
    ax4.tick_params(colors='black')
    ax4.legend(facecolor='white', labelcolor='black', fontsize=9, edgecolor='black')
    ax4.grid(True, alpha=0.2, color='#e0e0e0')

    plt.tight_layout(rect=[0, 0, 1, 0.96])
    plt.show()
    return df_result

# =============================================================================
# 6. LSTM TRAINING
# =============================================================================
def train_rnn(df, target_col, sequence_length=32, n_layers=2, units=64,
              plot_predictions=True, plot_start=None, plot_end=None):
    """
    Melatih LSTM dengan konfigurasi:
      - n_layers        : jumlah layer LSTM (1–4)
      - sequence_length : panjang sekuens input (32 / 64 / 128)
      - units           : jumlah unit per layer (64 / 128 / 256)

    LSTM memiliki dua state: hidden state (h_t) dan cell state (c_t).
    Cell state memungkinkan informasi mengalir tanpa modifikasi (long-term memory),
    sementara gate mekanisme (input/forget/output) mengontrol apa yang disimpan/dilupakan.
    """
    print(f"🤖 LSTM: {target_col} | L{n_layers}-S{sequence_length}-U{units}")

    all_num    = df.select_dtypes('number').columns.tolist()
    leak_cols  = LEAKAGE_MAP.get(target_col, [target_col])
    features_X = [c for c in all_num if c not in leak_cols]

    if target_col not in all_num:
        print(f"   ⚠️ '{target_col}' not in numeric columns — skipping.")
        return None
    if len(features_X) < 3:
        print(f"   ⚠️ Too few features — skipping.")
        return None

    print(f"   ✅ Features for X: {len(features_X)} cols | Excluded (leakage): {leak_cols}")
    if target_col == 'PR (%)':
        pr_in_X = 'PR (%)' in features_X
        print(f"   {'✅' if pr_in_X else '❌'} PR (%) in features_X: {pr_in_X}")

    X_raw = df[features_X].ffill().values
    y_raw = df[target_col].ffill().values
    ts    = df.index[sequence_length:]

    X_seq, y_seq = [], []
    for i in range(sequence_length, len(X_raw)):
        X_seq.append(X_raw[i - sequence_length:i])
        y_seq.append(y_raw[i])

    if len(X_seq) < 50:
        print(f"   ⚠️ Only {len(X_seq)} sequences — skipping.")
        return None

    X_arr = np.array(X_seq, dtype=np.float32)
    y_arr = np.array(y_seq, dtype=np.float32)

    split           = int(0.9 * len(X_arr))
    X_train, X_test = X_arr[:split], X_arr[split:]
    y_train, y_test = y_arr[:split], y_arr[split:]
    test_ts         = ts[split:]

    n_feat    = X_train.shape[2]
    const_idx = np.where(np.std(X_train.reshape(-1, n_feat), axis=0) == 0)[0]

    sc_X = MinMaxScaler()
    sc_y = MinMaxScaler()

    Xtr = sc_X.fit_transform(X_train.reshape(-1, n_feat)).reshape(X_train.shape)
    Xte = sc_X.transform(X_test.reshape(-1, n_feat)).reshape(X_test.shape)
    if len(const_idx):
        Xtr[:, :, const_idx] = 0.5
        Xte[:, :, const_idx] = 0.5

    ytr = sc_y.fit_transform(y_train.reshape(-1, 1)).flatten()
    yte = sc_y.transform(y_test.reshape(-1, 1)).flatten()

    # ── CHANGE 2: BUILD LSTM MODEL ───────────────────────────────────────────
    model = Sequential()

    # Layer pertama (selalu ada)
    model.add(LSTM(
        units,
        return_sequences=(n_layers > 1),
        input_shape=(sequence_length, len(features_X))
    ))
    model.add(Dropout(0.2))

    # Layer tengah (jika n_layers > 2)
    for i in range(1, n_layers - 1):
        model.add(LSTM(units, return_sequences=True))
        model.add(Dropout(0.2))

    # Layer terakhir (jika n_layers >= 2)
    if n_layers >= 2:
        model.add(LSTM(units, return_sequences=False))
        model.add(Dropout(0.2))

    model.add(Dense(32, activation='relu'))
    model.add(Dense(1))
    model.compile(optimizer=Adam(learning_rate=0.001), loss='mse')
    # ────────────────────────────────────────────────────────────────────────

    model.fit(
        Xtr, ytr,
        epochs=500,
        batch_size=32,
        validation_split=0.1111,
        callbacks=[EarlyStopping(patience=15, restore_best_weights=True)],
        verbose=0
    )

    y_pred_scaled = model.predict(Xte, verbose=0).flatten()
    y_pred        = sc_y.inverse_transform(y_pred_scaled.reshape(-1, 1)).flatten()
    y_test_orig   = sc_y.inverse_transform(yte.reshape(-1, 1)).flatten()

    rmse  = np.sqrt(mean_squared_error(y_test_orig, y_pred))
    r2    = r2_score(y_test_orig, y_pred)

    ymax  = np.max(np.abs(y_test_orig))
    nrmse = (rmse / ymax * 100) if ymax > 0 else np.nan

    mask  = y_test_orig > 1
    mape  = (np.mean(np.abs((y_test_orig[mask] - y_pred[mask]) / y_test_orig[mask])) * 100
             if mask.sum() > 0 else np.nan)

    print(f"✅ {target_col}: RMSE={rmse:.3f} | nRMSE(ymax)={nrmse:.1f}% | "
          f"MAPE={mape:.1f}% | R²={r2:.3f}")

    if plot_predictions:
        plot_date_filtered(y_test_orig, y_pred, test_ts, target_col,
                           rmse, nrmse, mape, r2,
                           n_layers, sequence_length, units,
                           plot_start=plot_start, plot_end=plot_end)

    return {
        'model'          : model,
        'rmse'           : rmse,
        'nrmse'          : nrmse,
        'mape'           : mape,
        'r2'             : r2,
        'X_test_scaled'  : Xte,
        'feature_names'  : features_X,
        'y_test_orig'    : y_test_orig,
        'y_pred'         : y_pred,
        'test_timestamps': test_ts,
        'scaler_y'       : sc_y,
    }

# =============================================================================
# 7. BEST-MODELS COMPARISON
# =============================================================================
def plot_best_models_comparison(best_models, plot_start=None, plot_end=None):
    n = len(best_models)
    if n == 0: return

    fig, axes = plt.subplots(2, n, figsize=(10 * n, 12))
    fig.patch.set_facecolor('white')
    if n == 1:
        axes = axes.reshape(2, 1)
    for ax in axes.flat:
        ax.set_facecolor('white')
        for sp in ax.spines.values():
            sp.set_edgecolor('black')

    # ── CHANGE 3: Update comparison title to LSTM ───────────────────────────
    fig.suptitle('🏆 BEST LSTM MODELS — Side-by-Side Comparison',
                 fontsize=18, fontweight='bold', color='black')

    for idx, (target, res) in enumerate(best_models.items()):
        df_r = pd.DataFrame(
            {'Actual': res['y_test_orig'], 'Prediction': res['y_pred']},
            index=pd.to_datetime(res['test_timestamps'])
        )
        df_r['Residual'] = df_r['Actual'] - df_r['Prediction']

        if plot_start and plot_end:
            mask    = (df_r.index >= plot_start) & (df_r.index <= plot_end)
            plot_df = df_r.loc[mask] if mask.any() else df_r
            wlabel  = f"[{plot_start[:10]} → {plot_end[:10]}]"
        else:
            plot_df = df_r
            wlabel  = "[Full Test Period]"

        t = plot_df.index
        a = plot_df['Actual'].values
        p = plot_df['Prediction'].values
        r = plot_df['Residual'].values

        axes[0, idx].plot(t, a, color='#4fc3f7', lw=2.5, label='Actual')
        axes[0, idx].plot(t, p, color='#ff7043', lw=2.5, label='Prediction', ls='--')
        axes[0, idx].fill_between(t, a, p, alpha=0.18, color='#ffd54f')
        axes[0, idx].set_title(f'{target}  |  R²={res["r2"]:.3f}\n{wlabel}',
                               color='black', fontweight='bold')
        axes[0, idx].tick_params(colors='black')
        axes[0, idx].xaxis.set_major_formatter(mdates.DateFormatter('%d-%b'))
        axes[0, idx].legend(facecolor='white', labelcolor='black', fontsize=9, edgecolor='black')
        axes[0, idx].grid(True, alpha=0.2, color='#e0e0e0')

        axes[1, idx].plot(t, r, color='#69f0ae', lw=2, alpha=0.8)
        axes[1, idx].axhline(0, color='black', lw=1.2, alpha=0.6)
        axes[1, idx].fill_between(t, r, 0, where=(r >= 0), alpha=0.22, color='#69f0ae')
        axes[1, idx].fill_between(t, r, 0, where=(r <  0), alpha=0.22, color='#ff7043')
        axes[1, idx].set_title(f'Residuals  |  RMSE={res["rmse"]:.3f}',
                               color='black', fontweight='bold')
        axes[1, idx].tick_params(colors='black')
        axes[1, idx].xaxis.set_major_formatter(mdates.DateFormatter('%d-%b'))
        axes[1, idx].grid(True, alpha=0.2, color='#e0e0e0')

    plt.tight_layout()
    plt.show()

# =============================================================================
# 8. EDA
# =============================================================================
def plot_eda(df):
    eff_cols = [c for c in df.columns if 'Efficiency' in c or 'PR' in c]
    fig, axes = plt.subplots(1, 2, figsize=(20, 7))
    fig.patch.set_facecolor('white')
    for ax in axes:
        ax.set_facecolor('white')

    corr = df.select_dtypes('number').corr()
    sns.heatmap(corr, annot=False, cmap='RdBu_r', center=0,
                ax=axes[0], linewidths=0.3, cbar_kws={'shrink': 0.8})
    axes[0].set_title('Correlation Matrix', color='black', fontweight='bold', fontsize=13)
    axes[0].tick_params(colors='black', labelsize=7)

    for col in eff_cols:
        if col in df.columns:
            axes[1].plot(df.index, df[col], lw=1.2, alpha=0.75, label=col)
    axes[1].set_title('All Efficiencies & PR Over Time', color='black', fontweight='bold', fontsize=13)
    axes[1].set_ylabel('Efficiency / PR (%)', color='black')
    axes[1].tick_params(colors='black')
    axes[1].legend(fontsize=8, facecolor='white', labelcolor='black', edgecolor='black')
    axes[1].grid(True, alpha=0.2, color='#e0e0e0')

    fig.suptitle('📊 Exploratory Data Analysis', fontsize=16, color='black', fontweight='bold')
    plt.tight_layout(); plt.show()

# =============================================================================
# 9. MAIN
# =============================================================================
def main():
    print("\n" + "=" * 80)
    # ── CHANGE 3: Update main banner to LSTM ────────────────────────────────
    print("🚀 HYBRID PLTS — LSTM")
    print(f"   Variasi: Layers={_LAYERS_LIST} | SeqLen={_SEQLEN_LIST} | Units={_UNITS_LIST}")
    print(f"   Total eksperimen per target: {len(EXPERIMENT_GRID)}")
    print(f"   Config: {BEST_CONFIG}")
    print("=" * 80)

    df = load_hybrid_data()
    if len(df) < 100:
        print("❌ Insufficient data!"); return None, None

    df = engineer_hybrid_features(df)
    df = clean_data(df)
    plot_eda(df)

    all_rows        = []
    best_per_target = {}

    for target in BEST_CONFIG['targets']:
        if target not in df.columns:
            print(f"⚠️ '{target}' not found — skipping."); continue

        best_r2, best_res = -np.inf, None
        all_xai_results   = {}

        print(f"\n{'─'*60}")
        print(f"🎯 Target: {target}  ({len(EXPERIMENT_GRID)} eksperimen)")
        print(f"{'─'*60}")

        for exp_no, (n_layers, seq_len, units) in enumerate(EXPERIMENT_GRID, 1):
            print(f"\n  [{exp_no:02d}/{len(EXPERIMENT_GRID)}]", end=" ")
            res = train_rnn(
                df, target,
                sequence_length  = seq_len,
                n_layers         = n_layers,
                units            = units,
                plot_predictions = BEST_CONFIG['plot_predictions'],
                plot_start       = BEST_CONFIG['plot_start'],
                plot_end         = BEST_CONFIG['plot_end'],
            )
            if res is None: continue

            all_rows.append({
                'Target'    : target,
                'Model'     : 'LSTM',                          # ← LSTM
                'Layers'    : n_layers,
                'SeqLen'    : seq_len,
                'Units'     : units,
                'RMSE'      : round(res['rmse'],  4),
                'nRMSE'     : round(res['nrmse'], 2),
                'MAPE'      : round(res['mape'],  2),
                'R2'        : round(res['r2'],    4),
            })

            target_r2s = [r['R2'] for r in all_rows if r['Target'] == target]
            if len(target_r2s) >= 2 and res['r2'] >= np.percentile(target_r2s, 80):
                xai_current = compute_xai(
                    res['model'], res['X_test_scaled'],
                    res['feature_names'], df, target
                )
                all_xai_results[f"L{n_layers}-S{seq_len}-U{units}"] = {
                    'xai_df': xai_current,
                    'r2'    : res['r2'],
                    'rmse'  : res['rmse'],
                }

            if res['r2'] > best_r2:
                best_r2  = res['r2']
                best_res = res
                best_per_target[target] = res

        if best_res:
            print(f"\n🏆 BEST {target}: R²={best_r2:.4f} "
                  f"(L{best_res.get('n_layers','?')}-"
                  f"S{best_res.get('seq_len','?')}-"
                  f"U{best_res.get('units','?')})")

            xai_df = compute_xai(
                best_res['model'], best_res['X_test_scaled'],
                best_res['feature_names'], df, target
            )

            plot_xai(xai_df, f"{target} — Best LSTM")
            plot_xai_scatter_features(xai_df, target, best_res, df)
            plot_xai_residual_by_quartile(xai_df, target, best_res, df)
            plot_xai_temporal_error(target, best_res)

            if all_xai_results:
                plot_xai_comparison(all_xai_results, target)

            if all_rows:
                summary_so_far = pd.DataFrame(all_rows)
                plot_config_impact_analysis(summary_so_far, target)

    if best_per_target:
        plot_best_models_comparison(best_per_target,
                                    plot_start=BEST_CONFIG['plot_start'],
                                    plot_end  =BEST_CONFIG['plot_end'])

    if all_rows:
        summary = pd.DataFrame(all_rows)
        summary_sorted = summary.sort_values(
            ['R2', 'nRMSE'], ascending=[False, True]
        )
        print("\n📊 ALL EXPERIMENTS — sorted by R² ↓, nRMSE ↑:")
        print(summary_sorted.to_string(index=False))
        print("\n📊 TOP 5 (R² ↓, nRMSE ↑):")
        print(summary_sorted.head(5).to_string(index=False))
    else:
        summary = pd.DataFrame()

    return summary, best_per_target


if __name__ == "__main__":
    results, best_models = main()
```
