```python
# ═══════════════════════════════════════════════════════════════════════════════
#  RNN vs LSTM vs GRU  —  6 Targets  |  v2
#  XAI : Permutation Importance  +  SHAP (GradientExplainer)
#          ↳ SHAP Bar  |  SHAP Beeswarm  |  SHAP Temporal Heatmap
#          ↳ SHAP Cross-Model Comparison  |  SHAP + Metrics Joint Dashboard
#  + Anomaly Score  +  Sequence Prediction Accuracy
#  + Composite Model Selection Dashboard
#  ─────────────────────────────────────────────────────────────────────────────
#  [NEW v2]
#    ► 9K  Epochs Effect   : GRU 1-layer (seq=32, units=256)  →  4-panel plot
#    ► 9L  Blocked CV      : Blok waktu sebagai fold           →  fold-box plot
#                            (Roberts 2017; Cerqueira 2019)
#    ► 9L  Rolling-Origin  : Expanding-window forward CV       →  idem
#                            (Schnaubelt 2020; Maldonado 2021)
#    ► 9M  Aggregate Val   : Error di level jam / hari         →  bar-chart
#                            (Kennedy et al. 2023)
#    ► 9N  Bootstrap CI    : Distribusi performa ≠ titik tunggal
#                            (Lehmann et al. 2025)
# ═══════════════════════════════════════════════════════════════════════════════

import os, sys, subprocess
os.environ['CUDA_VISIBLE_DEVICES'] = '-1'
os.environ['TF_CPP_MIN_LOG_LEVEL'] = '3'

# ── Auto-install SHAP if missing ──────────────────────────────────────────────
try:
    import shap
    print(f"✅ SHAP {shap.__version__} already installed")
except ImportError:
    print("📦 Installing SHAP …")
    subprocess.check_call([sys.executable, '-m', 'pip', 'install', 'shap', '-q'])
    import shap
    print(f"✅ SHAP {shap.__version__} installed")

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.dates as mdates
import matplotlib.gridspec as gridspec
import matplotlib.patches as mpatches
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')

from sklearn.preprocessing import MinMaxScaler
from sklearn.metrics import mean_squared_error, r2_score

# ── TensorFlow ────────────────────────────────────────────────────────────────
try:
    import tensorflow as tf
    from tensorflow.keras.models import Sequential
    from tensorflow.keras.layers import SimpleRNN, LSTM, GRU, Dense, Dropout
    from tensorflow.keras.optimizers import Adam
    from tensorflow.keras.callbacks import EarlyStopping, Callback
    print(f"✅ TensorFlow {tf.__version__} OK")
except Exception as e:
    print(f"❌ TensorFlow error: {e}"); sys.exit()

# ── White theme ───────────────────────────────────────────────────────────────
plt.style.use('default')
plt.rcParams.update({
    'figure.facecolor': 'white', 'axes.facecolor': 'white',
    'axes.edgecolor': '#cccccc', 'axes.labelcolor': '#222222',
    'xtick.color': '#444444', 'ytick.color': '#444444',
    'text.color': '#222222', 'legend.facecolor': 'white',
    'legend.edgecolor': '#cccccc', 'grid.color': '#eeeeee',
    'grid.alpha': 0.6, 'font.size': 10,
})

print("=" * 80)
print("  RNN vs LSTM vs GRU  |  v2  with CV + Aggregate Val + Epochs Analysis")
print("=" * 80)

# ─────────────────────────────────────────────────────────────────────────────
# CONFIGURATION  (original)
# ─────────────────────────────────────────────────────────────────────────────
MODEL_CONFIGS = {
    'RNN':  {'seq': 32, 'units': 128, 'color': '#e63946', 'marker': 'o'},
    'LSTM': {'seq': 32, 'units': 128, 'color': '#2a9d8f', 'marker': 's'},
    'GRU':  {'seq': 32, 'units': 256, 'color': '#f4a261', 'marker': '^'},
}

TRAIN_CFG = {
    'epochs': 300, 'batch_size': 32, 'dropout': 0.2,
    'system_capacity_wp': 2250, 'panel_area_m2': 10.893,
}

TARGETS = [
    'Power PV', 'PR (%)', 'Efficiency PV (%)',
    'Efficiency Temp (%)', 'Efficiency Battery (%)', 'Efficiency Inv (%)',
]

LEAKAGE_MAP = {
    'Power PV':               ['Power PV', 'Efficiency PV (%)',
                               'Efficiency System (%)', 'PR (%)', 'P_pv'],
    'PR (%)':                 [],
    'Efficiency PV (%)':      ['Power PV', 'Efficiency PV (%)'],
    'Efficiency Temp (%)':    [],
    'Efficiency Battery (%)': [],
    'Efficiency Inv (%)':     ['Efficiency Inv (%)', 'P_ac'],
}

XAI_N_PERM    = 10
SHAP_BG       = 80
SHAP_EXPLAIN  = 100
SHAP_TOP_N    = 12
ANOMALY_THRESH = 2.5
SEQ_TOL       = 0.10

# ─────────────────────────────────────────────────────────────────────────────
# [NEW v2]  CONFIGURATION — Epochs Effect + CV + Aggregate + Bootstrap
# ─────────────────────────────────────────────────────────────────────────────

# -- 9K: Epochs Effect ---------------------------------------------------------
EPOCH_CHECKPOINTS  = [10, 25, 50, 75, 100, 150, 200, 250, 300]
EPOCH_ANALYSIS_TGT = 'Power PV'          # target for epochs analysis
GRU_EPOCH_SEQ      = 32                  # fixed GRU 1-layer spec
GRU_EPOCH_UNITS    = 256

# -- 9L: Time-Series Cross-Validation ------------------------------------------
#   CV_N_FOLDS = 3 is intentionally conservative (108 trainings total).
#   Increase to 5 on GPU; reduce to 2 on limited hardware.
CV_N_FOLDS        = 3
CV_MIN_TRAIN_RATIO = 0.50  # rolling origin: min share of data for first train
CV_GAP            = 0      # optional gap (samples) between train and val fold
CV_PATIENCE       = 10     # EarlyStopping patience inside CV (shorter than main)

# -- 9M: Aggregate Validation --------------------------------------------------
AGG_FREQS = [('1h', 'Hourly'), ('1D', 'Daily')]

# -- 9N: Bootstrap Distributional Uncertainty ----------------------------------
BOOTSTRAP_N  = 500          # resamples (500 balances accuracy vs speed)
BOOTSTRAP_CI = 0.95         # confidence level


# ─────────────────────────────────────────────────────────────────────────────
# 1.  DATA LOADING
# ─────────────────────────────────────────────────────────────────────────────
def load_hybrid_data(env_file='/content/Lingkungan.csv',
                     dc_file='/content/Elektrik DC.csv',
                     ac_file='/content/Elektrik AC.csv'):
    print("\n📁 Loading 3 CSV files …")
    df_env = pd.read_csv(env_file)
    df_dc  = pd.read_csv(dc_file)
    df_ac  = pd.read_csv(ac_file)

    def parse_dt(df):
        df = df.copy()
        df['datetime'] = pd.to_datetime(
            df['Date'].astype(str) + ' ' + df['Time'].astype(str), errors='coerce')
        return df.dropna(subset=['datetime']).sort_values('datetime')

    df_env = parse_dt(df_env).set_index('datetime')
    df_dc  = parse_dt(df_dc ).set_index('datetime')
    df_ac  = parse_dt(df_ac ).set_index('datetime')

    tol = pd.Timedelta('10m')
    df  = pd.merge_asof(df_dc.sort_index(), df_env.sort_index(),
                        left_index=True, right_index=True,
                        tolerance=tol, direction='nearest')
    df  = pd.merge_asof(df.sort_index(), df_ac.sort_index(),
                        left_index=True, right_index=True,
                        tolerance=tol, direction='nearest')

    df['hour']       = df.index.hour
    df['is_daytime'] = ((df.index.hour >= 6) & (df.index.hour <= 18)).astype(int)
    df['hour_sin']   = np.sin(2 * np.pi * df.index.hour / 24)
    df['hour_cos']   = np.cos(2 * np.pi * df.index.hour / 24)

    print(f"✅ Merged: {len(df)} rows, {df.shape[1]} columns")
    return df


# ─────────────────────────────────────────────────────────────────────────────
# 2.  FEATURE ENGINEERING
# ─────────────────────────────────────────────────────────────────────────────
def engineer_features(df, system_capacity_wp=2250, panel_area_m2=10.893):
    df = df.copy()
    for col in df.columns:
        df[col] = pd.to_numeric(df[col], errors='coerce')

    df['delta_t'] = df.index.to_series().diff().dt.total_seconds() / 3600
    df['P_ac']    = df['V AC'] * df['I AC']
    df['E_ac']    = df['P_ac'] * df['delta_t']

    if 'Irradiance' in df.columns and 'Power PV' in df.columns:
        df['Efficiency PV (%)'] = (
            df['Power PV'] / (df['Irradiance'] * panel_area_m2)) * 100

    df['P_batt']      = df['V25 (V)'] * df['I ACS758 (A)']
    df['E_batt']      = df['P_batt'] * df['delta_t']
    df['is_charging'] = (
        (df.index.hour >= 6) & (df.index.hour < 18) &
        (df['V25 (V)'].fillna(0) > 0) &
        (df['I ACS758 (A)'].fillna(0) > 0)
    )
    df['E_charge']    = np.where(df['is_charging'],  df['E_batt'], np.nan)
    df['E_discharge'] = np.where(~df['is_charging'], df['E_batt'], np.nan)

    df['E_charge_total']    = (df['E_charge'].fillna(0)
                                .rolling('24h', min_periods=1).sum().shift(1))
    df['E_discharge_total'] = (df['E_discharge'].fillna(0)
                                .rolling('24h', min_periods=1).sum().shift(1))
    df['Efficiency Battery (%)'] = (
        df['E_discharge_total'] / df['E_charge_total'].replace(0, np.nan)) * 100

    df['P_pv']           = df['V220 (V)'] * df['I ACS712 (A)']
    df['E_dc_total']     = df['P_pv'] * df['delta_t']
    df['E_dc_total_24h'] = (df['E_dc_total']
                             .rolling('24h', min_periods=1).sum().shift(1))
    df['E_ac_24h']       = df['E_ac'].rolling('24h', min_periods=1).sum().shift(1)
    df['Efficiency Inv (%)'] = (df['E_ac_24h'] / df['E_dc_total_24h']) * 100
    df['Efficiency Inv (%)'] = df['Efficiency Inv (%)'].replace([np.inf, -np.inf], np.nan)

    df['Efficiency Temp (%)'] = (
        1 - 0.004 * (df['Temp PV (C)'].fillna(25) - 25)) * 100

    df['Irrad_kWh']       = (df['Irradiance'] * df['delta_t']) / 1000
    df['Irrad_total_24h'] = (df['Irrad_kWh']
                              .rolling('24h', min_periods=1).sum().shift(1))
    df['PR (%)'] = (
        df['E_ac_24h'] / (system_capacity_wp * df['Irrad_total_24h'])) * 100

    df['Efficiency PV (%)']       = df['Efficiency PV (%)'].clip(0, 25)
    df['Efficiency Battery (%)']  = df['Efficiency Battery (%)'].clip(0, 100)
    df['Efficiency Inv (%)']      = df['Efficiency Inv (%)'].clip(0, 100)
    df['Efficiency Temp (%)']     = df['Efficiency Temp (%)'].clip(50, 100)
    df['PR (%)']                  = df['PR (%)'].clip(0, 100)

    print("✅ Feature engineering done — 6 targets ready")
    return df


def clean_data(df):
    df = df.copy()
    for col in df.select_dtypes(include=[np.number]).columns:
        df[col] = df[col].fillna(df[col].median())
        df[col] = np.nan_to_num(df[col])
        q1, q3 = df[col].quantile([0.25, 0.75])
        df[col] = df[col].clip(q1 - 1.5*(q3-q1), q3 + 1.5*(q3-q1))
    return df


# ─────────────────────────────────────────────────────────────────────────────
# 3.  BUILD MODEL
# ─────────────────────────────────────────────────────────────────────────────
def build_model(arch, seq_len, n_features, units, dropout=0.2):
    m = Sequential(name=arch)
    if arch == 'RNN':
        m.add(SimpleRNN(units, input_shape=(seq_len, n_features),
                        return_sequences=False))
    elif arch == 'LSTM':
        m.add(LSTM(units, input_shape=(seq_len, n_features),
                   return_sequences=False))
    elif arch == 'GRU':
        m.add(GRU(units, input_shape=(seq_len, n_features),
                  return_sequences=False))
    m.add(Dropout(dropout))
    m.add(Dense(32, activation='relu'))
    m.add(Dense(1))
    m.compile(optimizer=Adam(learning_rate=0.001), loss='mse')
    return m


# ─────────────────────────────────────────────────────────────────────────────
# 4A.  XAI — PERMUTATION FEATURE IMPORTANCE
# ─────────────────────────────────────────────────────────────────────────────
def compute_permutation_importance(model, Xte, y_test_orig, sc_y,
                                   feature_names, n_repeats=XAI_N_PERM):
    y_base   = sc_y.inverse_transform(model.predict(Xte, verbose=0)).flatten()
    base_mse = mean_squared_error(y_test_orig, y_base)

    imp = np.zeros(len(feature_names))
    for fi in range(len(feature_names)):
        deltas = []
        for _ in range(n_repeats):
            Xp = Xte.copy()
            Xp[:, :, fi] = Xp[np.random.permutation(len(Xp)), :, fi]
            yp = sc_y.inverse_transform(model.predict(Xp, verbose=0)).flatten()
            deltas.append(mean_squared_error(y_test_orig, yp) - base_mse)
        imp[fi] = np.mean(deltas)

    imp = np.clip(imp, 0, None)
    tot = imp.sum()
    if tot > 0:
        imp /= tot
    return imp


# ─────────────────────────────────────────────────────────────────────────────
# 4B.  XAI — SHAP  (GradientExplainer, fallback KernelExplainer)
# ─────────────────────────────────────────────────────────────────────────────
def compute_shap(model, Xtr, Xte, feature_names,
                 bg=SHAP_BG, n_explain=SHAP_EXPLAIN):
    print(f"      🔮 SHAP GradientExplainer  bg={bg}  explain={n_explain} …")

    bg_idx  = np.random.choice(len(Xtr), size=min(bg, len(Xtr)), replace=False)
    X_bg    = Xtr[bg_idx].astype(np.float32)
    ex_idx  = np.random.choice(len(Xte), size=min(n_explain, len(Xte)), replace=False)
    X_exp   = Xte[ex_idx].astype(np.float32)

    try:
        explainer = shap.GradientExplainer(model, X_bg)
        sv        = np.array(explainer.shap_values(X_exp))

        if isinstance(sv, list):
            sv = sv[0]
        while sv.ndim > 3:
            sv = sv.squeeze(-1)

        shap_3d  = sv
        shap_2d  = np.abs(sv).mean(axis=1)
        X_exp_2d = X_exp.mean(axis=1)

        print(f"         ✅ GradientExplainer done — shape {shap_2d.shape}")
        return shap_2d, shap_3d, X_exp_2d

    except Exception as e:
        print(f"         ⚠️  GradientExplainer failed ({e})")
        print("         ↳  Trying KernelExplainer (slower) …")
        try:
            n_s, seq, nf = X_bg.shape
            X_bg_2d       = X_bg.reshape(n_s, seq * nf)
            X_exp_2d_flat = X_exp.reshape(len(X_exp), seq * nf)

            def model_fn(x2d):
                return model.predict(
                    x2d.reshape(-1, seq, nf).astype(np.float32),
                    verbose=0).flatten()

            ke  = shap.KernelExplainer(model_fn, shap.sample(X_bg_2d, 30))
            sv2 = ke.shap_values(X_exp_2d_flat, nsamples=50, silent=True)
            sv3 = np.array(sv2).reshape(-1, seq, nf)
            s2d = np.abs(sv3).mean(axis=1)
            x2d = X_exp.mean(axis=1)
            print(f"         ✅ KernelExplainer done — shape {s2d.shape}")
            return s2d, sv3, x2d
        except Exception as e2:
            print(f"         ❌ KernelExplainer also failed: {e2}")
            return None, None, None


# ─────────────────────────────────────────────────────────────────────────────
# 5.  ANOMALY SCORE
# ─────────────────────────────────────────────────────────────────────────────
def anomaly_score(y_actual, y_pred, thresh=ANOMALY_THRESH):
    res   = y_actual - y_pred
    mu, sigma = res.mean(), res.std()
    if sigma == 0:
        sigma = 1e-9
    z     = (res - mu) / sigma
    flags = np.abs(z) > thresh
    return {
        'flags':           flags,
        'z_scores':        z,
        'anomaly_rate':    flags.mean() * 100,
        'mean_abs_zscore': np.abs(z).mean(),
        'n_anomalies':     int(flags.sum()),
    }


# ─────────────────────────────────────────────────────────────────────────────
# 6.  SEQUENCE PREDICTION ACCURACY
# ─────────────────────────────────────────────────────────────────────────────
def seq_accuracy(y_actual, y_pred, tol_pct=SEQ_TOL):
    tol = np.maximum(np.abs(y_actual) * tol_pct, 1e-6)
    tba = (np.abs(y_actual - y_pred) <= tol).mean() * 100
    da  = (np.sign(np.diff(y_actual)) == np.sign(np.diff(y_pred))).mean() * 100
    return {'tba': tba, 'da': da, 'seq_score': (tba + da) / 2}


# ─────────────────────────────────────────────────────────────────────────────
# 7.  TRAIN ONE ARCH × ONE TARGET
# ─────────────────────────────────────────────────────────────────────────────
def train_model(df, target_col, arch):
    cfg     = MODEL_CONFIGS[arch]
    seq_len = cfg['seq']

    all_num    = df.select_dtypes('number').columns.tolist()
    leak       = LEAKAGE_MAP.get(target_col, [target_col])
    features_X = [c for c in all_num if c not in leak]

    if target_col not in df.columns or len(features_X) < 3:
        print(f"   ⚠️  {arch}/{target_col}: skipped"); return None

    X_raw = df[features_X].ffill().values
    y_raw = df[target_col].ffill().values
    ts    = df.index[seq_len:]

    X_seq, y_seq = [], []
    for i in range(seq_len, len(X_raw)):
        X_seq.append(X_raw[i-seq_len:i])
        y_seq.append(y_raw[i])
    if len(X_seq) < 100:
        print(f"   ⚠️  {arch}/{target_col}: not enough sequences"); return None

    X_arr = np.array(X_seq, dtype=np.float32)
    y_arr = np.array(y_seq, dtype=np.float32)
    split = int(0.9 * len(X_arr))
    Xtr, Xte = X_arr[:split], X_arr[split:]
    ytr, yte = y_arr[:split], y_arr[split:]
    test_ts   = ts[split:]

    nf   = Xtr.shape[2]
    sc_X = MinMaxScaler(); sc_y = MinMaxScaler()
    Xtr  = sc_X.fit_transform(Xtr.reshape(-1, nf)).reshape(Xtr.shape)
    Xte  = sc_X.transform(Xte.reshape(-1, nf)).reshape(Xte.shape)
    ytr  = sc_y.fit_transform(ytr.reshape(-1, 1)).flatten()
    yte  = sc_y.transform(yte.reshape(-1, 1)).flatten()

    model = build_model(arch, seq_len, nf, cfg['units'], TRAIN_CFG['dropout'])
    model.fit(Xtr, ytr,
              epochs=TRAIN_CFG['epochs'],
              batch_size=TRAIN_CFG['batch_size'],
              validation_split=0.1,
              callbacks=[EarlyStopping(patience=20, restore_best_weights=True)],
              verbose=0)

    y_pred      = sc_y.inverse_transform(
                    model.predict(Xte, verbose=0)).flatten()
    y_test_orig = sc_y.inverse_transform(yte.reshape(-1, 1)).flatten()

    rmse  = np.sqrt(mean_squared_error(y_test_orig, y_pred))
    r2    = r2_score(y_test_orig, y_pred)
    ymax  = np.max(np.abs(y_test_orig))
    nrmse = (rmse / ymax * 100) if ymax > 0 else np.nan
    mask  = y_test_orig > 1
    mape  = (np.mean(np.abs((y_test_orig[mask] - y_pred[mask]) /
              y_test_orig[mask])) * 100) if mask.sum() > 0 else np.nan

    print(f"      🔍 Permutation importance ({arch}/{target_col}) …")
    perm_imp = compute_permutation_importance(
        model, Xte, y_test_orig, sc_y, features_X)

    shap_2d, shap_3d, X_exp_2d = compute_shap(model, Xtr, Xte, features_X)

    anom  = anomaly_score(y_test_orig, y_pred)
    s_acc = seq_accuracy(y_test_orig, y_pred)

    print(f"   ✅ {arch}/{target_col}  "
          f"nRMSE={nrmse:.1f}%  R²={r2:.3f}  "
          f"Anomaly={anom['anomaly_rate']:.1f}%  "
          f"SeqAcc={s_acc['seq_score']:.1f}%")

    return {
        'arch': arch, 'target': target_col,
        'model': model, 'feature_names': features_X,
        'scaler_x': sc_X, 'scaler_y': sc_y,
        'X_train_scaled': Xtr, 'X_test_scaled': Xte,
        'y_test_orig': y_test_orig, 'y_pred': y_pred,
        'test_timestamps': test_ts,
        'rmse': rmse, 'nrmse': nrmse, 'mape': mape, 'r2': r2,
        'perm_imp': perm_imp,
        'shap_2d': shap_2d, 'shap_3d': shap_3d, 'X_exp_2d': X_exp_2d,
        'anomaly': anom,
        'seq_acc': s_acc,
    }


# ─────────────────────────────────────────────────────────────────────────────
# 8.  COMPOSITE SCORE  (lower = better)
# ─────────────────────────────────────────────────────────────────────────────
def composite(res):
    nr  = res['nrmse'] if not np.isnan(res['nrmse']) else 100.0
    r2p = (1 - res['r2']) * 100
    an  = res['anomaly']['anomaly_rate']
    sp  = 100 - res['seq_acc']['seq_score']
    return 0.35*nr + 0.25*r2p + 0.25*an + 0.15*sp


# ═══════════════════════════════════════════════════════════════════════════════
#  [NEW v2]  SECTION A — EPOCHS EFFECT  (GRU 1-layer, seq=32, units=256)
# ═══════════════════════════════════════════════════════════════════════════════

class EpochCheckpointCallback(Callback):
    """
    Keras callback that evaluates test-set metrics at specific epoch
    checkpoints during a single training run.  Avoids retraining the model
    from scratch for each epoch count.
    """
    def __init__(self, X_test, y_test_orig, sc_y, checkpoints):
        super().__init__()
        self.X_test      = X_test.astype(np.float32)
        self.y_test_orig = y_test_orig
        self.sc_y        = sc_y
        self.checkpoints = set(checkpoints)
        self.records     = []   # filled on-the-fly

    def on_epoch_end(self, epoch, logs=None):
        ep = epoch + 1          # 1-indexed
        if ep not in self.checkpoints:
            return

        y_pred = self.sc_y.inverse_transform(
            self.model.predict(self.X_test, verbose=0)
        ).flatten()

        rmse = np.sqrt(mean_squared_error(self.y_test_orig, y_pred))
        r2   = r2_score(self.y_test_orig, y_pred)
        ymax = np.max(np.abs(self.y_test_orig))
        nrmse = (rmse / ymax * 100) if ymax > 0 else np.nan
        mask  = self.y_test_orig > 1
        mape  = (np.mean(np.abs(
            (self.y_test_orig[mask] - y_pred[mask]) / self.y_test_orig[mask]
        )) * 100) if mask.sum() > 0 else 0.0

        self.records.append({
            'epoch':      ep,
            'train_loss': float(logs.get('loss',     np.nan)),
            'val_loss':   float(logs.get('val_loss', np.nan)),
            'rmse': rmse, 'nrmse': nrmse, 'r2': r2,
            'mape': float(mape) if not np.isnan(mape) else 0.0,
        })


def analyze_epochs_effect(df):
    """
    Train GRU (1 layer, seq=32, units=256) once for max(EPOCH_CHECKPOINTS)
    epochs — without EarlyStopping — and record test metrics at each
    EPOCH_CHECKPOINT via EpochCheckpointCallback.

    Returns (epoch_records, full_history_dict).
    """
    print(f"\n📊 [A] Epochs Effect Analysis — GRU 1-layer "
          f"(seq={GRU_EPOCH_SEQ}, units={GRU_EPOCH_UNITS})  "
          f"target='{EPOCH_ANALYSIS_TGT}'")

    target_col = EPOCH_ANALYSIS_TGT
    seq_len    = GRU_EPOCH_SEQ
    units      = GRU_EPOCH_UNITS

    all_num    = df.select_dtypes('number').columns.tolist()
    leak       = LEAKAGE_MAP.get(target_col, [target_col])
    features_X = [c for c in all_num if c not in leak]

    if target_col not in df.columns or len(features_X) < 3:
        print(f"   ⚠️  '{target_col}' not found — skipping epochs analysis")
        return None, None

    X_raw = df[features_X].ffill().values
    y_raw = df[target_col].ffill().values

    X_seq, y_seq = [], []
    for i in range(seq_len, len(X_raw)):
        X_seq.append(X_raw[i-seq_len:i])
        y_seq.append(y_raw[i])

    if len(X_seq) < 100:
        print("   ⚠️  Too few sequences"); return None, None

    X_arr = np.array(X_seq, dtype=np.float32)
    y_arr = np.array(y_seq, dtype=np.float32)
    split = int(0.9 * len(X_arr))
    Xtr, Xte = X_arr[:split], X_arr[split:]
    ytr, yte  = y_arr[:split], y_arr[split:]

    nf   = Xtr.shape[2]
    sc_X = MinMaxScaler(); sc_y = MinMaxScaler()
    Xtr_s = sc_X.fit_transform(Xtr.reshape(-1, nf)).reshape(Xtr.shape)
    Xte_s = sc_X.transform(Xte.reshape(-1, nf)).reshape(Xte.shape)
    ytr_s = sc_y.fit_transform(ytr.reshape(-1, 1)).flatten()
    y_test_orig = sc_y.inverse_transform(yte.reshape(-1, 1)).flatten()

    model = build_model('GRU', seq_len, nf, units, TRAIN_CFG['dropout'])
    cb    = EpochCheckpointCallback(Xte_s, y_test_orig, sc_y, EPOCH_CHECKPOINTS)

    max_ep = max(EPOCH_CHECKPOINTS)
    print(f"   ⏳ Training for {max_ep} epochs (no EarlyStopping) …")
    hist = model.fit(
        Xtr_s, ytr_s,
        epochs=max_ep,
        batch_size=TRAIN_CFG['batch_size'],
        validation_split=0.1,
        callbacks=[cb],          # ← NO EarlyStopping here
        verbose=0
    )
    print(f"   ✅ {len(cb.records)} checkpoint records collected")

    full_history = {
        'loss':     hist.history.get('loss',     []),
        'val_loss': hist.history.get('val_loss', []),
    }
    del model
    return cb.records, full_history


# ═══════════════════════════════════════════════════════════════════════════════
#  [NEW v2]  SECTION B — TIME-SERIES CROSS-VALIDATION
# ═══════════════════════════════════════════════════════════════════════════════

def blocked_timesplit_splits(n, n_folds=CV_N_FOLDS, gap=CV_GAP):
    """
    Blocked Time-Split CV (Roberts et al. 2017; Cerqueira et al. 2019).

    Divides the time series into (n_folds+1) equal blocks.
    Fold k uses an *expanding* training window (blocks 0..k) and
    validates on block k+1.  Temporal order is strictly preserved.
    An optional 'gap' removes samples between train end and val start
    to avoid look-ahead from overlapping sequence windows.

    Returns list of (train_idx, val_idx) arrays.
    """
    block_size = n // (n_folds + 1)
    splits = []
    for fold in range(n_folds):
        train_end = (fold + 1) * block_size
        val_start = train_end + gap
        val_end   = min(val_start + block_size, n)
        if val_start >= val_end:
            continue
        splits.append((np.arange(0, train_end),
                       np.arange(val_start, val_end)))
    return splits


def rolling_origin_splits(n, n_folds=CV_N_FOLDS,
                           min_ratio=CV_MIN_TRAIN_RATIO, gap=CV_GAP):
    """
    Rolling-Origin / Forward CV — expanding window
    (Schnaubelt 2020; Maldonado & López 2021).

    Starts with at least min_ratio*n training samples, then expands
    the training window by one step per fold.  The validation window
    is always *after* the training window (+ optional gap).
    Suitable for non-stationary series or series with concept drift.

    Returns list of (train_idx, val_idx) arrays.
    """
    min_train = int(n * min_ratio)
    remaining = n - min_train
    step      = max(remaining // n_folds, 1)
    splits    = []
    for fold in range(n_folds):
        train_end = min_train + fold * step
        val_start = train_end + gap
        val_end   = min(val_start + step, n)
        if train_end >= n or val_start >= val_end:
            break
        splits.append((np.arange(0, train_end),
                       np.arange(val_start, val_end)))
    return splits


def _cv_single_fold(X_arr, y_arr, tr_idx, va_idx,
                    arch, seq_len, units, fold_num):
    """
    Train one model on tr_idx, evaluate on va_idx.
    Returns metrics dict or None if data insufficient.
    """
    Xtr, Xte = X_arr[tr_idx], X_arr[va_idx]
    ytr, yte  = y_arr[tr_idx], y_arr[va_idx]

    if len(Xtr) < 50 or len(Xte) < 5:
        return None

    nf   = Xtr.shape[2]
    sc_X = MinMaxScaler(); sc_y = MinMaxScaler()
    Xtr_s = sc_X.fit_transform(Xtr.reshape(-1, nf)).reshape(Xtr.shape)
    Xte_s = sc_X.transform(Xte.reshape(-1, nf)).reshape(Xte.shape)
    ytr_s = sc_y.fit_transform(ytr.reshape(-1, 1)).flatten()

    # Use smaller val_split only if enough training samples
    val_split = 0.1 if len(Xtr) >= 100 else 0.0

    model = build_model(arch, seq_len, nf, units, TRAIN_CFG['dropout'])
    model.fit(
        Xtr_s, ytr_s,
        epochs=TRAIN_CFG['epochs'],
        batch_size=TRAIN_CFG['batch_size'],
        validation_split=val_split,
        callbacks=[EarlyStopping(patience=CV_PATIENCE,
                                  restore_best_weights=True)],
        verbose=0,
    )

    y_pred = sc_y.inverse_transform(
        model.predict(Xte_s, verbose=0)).flatten()

    rmse  = np.sqrt(mean_squared_error(yte, y_pred))
    r2    = r2_score(yte, y_pred)
    ymax  = np.max(np.abs(yte))
    nrmse = (rmse / ymax * 100) if ymax > 0 else np.nan
    mask  = yte > 1
    mape  = (np.mean(np.abs((yte[mask] - y_pred[mask]) /
              yte[mask])) * 100) if mask.sum() > 0 else np.nan

    del model
    return {
        'fold': fold_num,
        'n_train': len(tr_idx), 'n_val': len(va_idx),
        'rmse': rmse, 'nrmse': nrmse, 'r2': r2,
        'mape': float(mape) if not np.isnan(mape) else np.nan,
    }


def run_timeseries_cv(df, n_folds=CV_N_FOLDS):
    """
    Run both Blocked and Rolling-Origin CV for ALL 3 architectures × 6 targets.

    NOTE: This trains n_folds × 2 strategies × 3 archs × 6 targets models.
    On CPU (~108 trainings) expect 30-90 min.  On GPU: ~5-15 min.

    Returns:
        cv_results['blocked'][arch][target] = [fold_metric_dict, ...]
        cv_results['rolling'][arch][target] = [fold_metric_dict, ...]
    """
    print(f"\n🔁 [B] Time-Series Cross-Validation  "
          f"({n_folds} folds × 2 strategies × "
          f"{len(MODEL_CONFIGS)} archs × {len(TARGETS)} targets = "
          f"{n_folds*2*len(MODEL_CONFIGS)*len(TARGETS)} trainings)")
    print("   Blocked CV  : Roberts 2017 / Cerqueira 2019")
    print("   Rolling-Origin CV : Schnaubelt 2020 / Maldonado 2021")
    print("─" * 72)

    cv_results = {
        'blocked': {a: {t: [] for t in TARGETS} for a in MODEL_CONFIGS},
        'rolling': {a: {t: [] for t in TARGETS} for a in MODEL_CONFIGS},
    }

    for tgt in TARGETS:
        all_num    = df.select_dtypes('number').columns.tolist()
        leak       = LEAKAGE_MAP.get(tgt, [tgt])
        features_X = [c for c in all_num if c not in leak]

        if tgt not in df.columns or len(features_X) < 3:
            continue

        for arch in MODEL_CONFIGS:
            cfg     = MODEL_CONFIGS[arch]
            seq_len = cfg['seq']
            units   = cfg['units']

            X_raw = df[features_X].ffill().values
            y_raw = df[tgt].ffill().values

            X_seq, y_seq = [], []
            for i in range(seq_len, len(X_raw)):
                X_seq.append(X_raw[i-seq_len:i])
                y_seq.append(y_raw[i])

            if len(X_seq) < 100:
                continue

            X_arr = np.array(X_seq, dtype=np.float32)
            y_arr = np.array(y_seq, dtype=np.float32)
            n     = len(X_arr)

            # -- Blocked CV --
            b_splits = blocked_timesplit_splits(n, n_folds)
            for fi, (tr, va) in enumerate(b_splits):
                r = _cv_single_fold(X_arr, y_arr, tr, va,
                                    arch, seq_len, units, fi+1)
                if r:
                    cv_results['blocked'][arch][tgt].append(r)

            # -- Rolling-Origin CV --
            r_splits = rolling_origin_splits(n, n_folds)
            for fi, (tr, va) in enumerate(r_splits):
                r = _cv_single_fold(X_arr, y_arr, tr, va,
                                    arch, seq_len, units, fi+1)
                if r:
                    cv_results['rolling'][arch][tgt].append(r)

            b_nrmse = np.mean([f['nrmse'] for f in cv_results['blocked'][arch][tgt]
                               if not np.isnan(f['nrmse'])])
            r_nrmse = np.mean([f['nrmse'] for f in cv_results['rolling'][arch][tgt]
                               if not np.isnan(f['nrmse'])])
            print(f"   ✅ {arch}/{tgt:28s}  "
                  f"Blocked nRMSE={b_nrmse:.1f}%  "
                  f"Rolling nRMSE={r_nrmse:.1f}%")

    return cv_results


# ═══════════════════════════════════════════════════════════════════════════════
#  [NEW v2]  SECTION C — AGGREGATE VALIDATION  +  BOOTSTRAP CI
# ═══════════════════════════════════════════════════════════════════════════════

def aggregate_validate(rba):
    """
    Validate predictions at aggregate temporal levels (hourly, daily).
    Kennedy et al. 2023: when the target of inference is an aggregate
    (group mean, daily total), the validation error must be measured
    at that aggregate level, not at the raw sample level.

    Returns: agg_results[arch][target] = {
        'point_rmse', 'point_r2',
        'hourly_rmse', 'hourly_r2',
        'daily_rmse',  'daily_r2'
    }
    """
    print("\n📐 [C] Aggregate Validation (Kennedy et al. 2023) …")
    agg_results = {}

    for arch in MODEL_CONFIGS:
        agg_results[arch] = {}
        for tgt in TARGETS:
            res = rba[arch].get(tgt)
            if res is None:
                agg_results[arch][tgt] = {}
                continue

            ts     = pd.to_datetime(res['test_timestamps'])
            actual = res['y_test_orig']
            pred   = res['y_pred']

            df_p = pd.DataFrame(
                {'actual': actual, 'pred': pred}, index=ts
            ).dropna()

            out = {}

            # Point-level
            if len(df_p) >= 2:
                out['point_rmse'] = np.sqrt(
                    mean_squared_error(df_p['actual'], df_p['pred']))
                out['point_r2'] = r2_score(df_p['actual'], df_p['pred'])

            # Aggregate levels
            for freq, label in AGG_FREQS:
                try:
                    df_agg = df_p.resample(freq).mean().dropna()
                    if len(df_agg) >= 2:
                        key = 'hourly' if freq == '1h' else 'daily'
                        out[f'{key}_rmse'] = np.sqrt(
                            mean_squared_error(df_agg['actual'], df_agg['pred']))
                        out[f'{key}_r2'] = r2_score(df_agg['actual'], df_agg['pred'])
                except Exception:
                    pass

            agg_results[arch][tgt] = out

    print("   ✅ Aggregate validation complete")
    return agg_results


def bootstrap_metric_ci(y_actual, y_pred,
                         metric='rmse',
                         n_boot=BOOTSTRAP_N, ci=BOOTSTRAP_CI):
    """
    Parametric-free bootstrap CI for RMSE or R².
    Lehmann et al. 2025: report distributional uncertainty instead
    of single-point estimates.

    Returns (bootstrap_mean, lower_bound, upper_bound).
    """
    n = len(y_actual)
    boot_vals = []
    for _ in range(n_boot):
        idx  = np.random.choice(n, size=n, replace=True)
        ya_b = y_actual[idx]; yp_b = y_pred[idx]
        if metric == 'rmse':
            boot_vals.append(np.sqrt(mean_squared_error(ya_b, yp_b)))
        elif metric == 'r2':
            if np.std(ya_b) > 0:
                boot_vals.append(r2_score(ya_b, yp_b))

    if not boot_vals:
        return (np.nan, np.nan, np.nan)

    alpha = (1 - ci) / 2
    mean_v = np.mean(boot_vals)
    lower  = np.percentile(boot_vals, alpha * 100)
    upper  = np.percentile(boot_vals, (1 - alpha) * 100)
    return (mean_v, lower, upper)


def compute_performance_distributions(rba):
    """
    Compute bootstrap CIs (RMSE and R²) for all arch × target combinations.
    Returns nested dict: dist[arch][target] = {'rmse': (m,lo,hi), 'r2': ...}
    """
    print(f"\n🔵 [C] Bootstrap CI  "
          f"(N={BOOTSTRAP_N}, CI={int(BOOTSTRAP_CI*100)}%)  …")
    dist_results = {}

    for arch in MODEL_CONFIGS:
        dist_results[arch] = {}
        for tgt in TARGETS:
            res = rba[arch].get(tgt)
            if res is None:
                dist_results[arch][tgt] = {}
                continue

            ya = res['y_test_orig']; yp = res['y_pred']
            rmse_ci = bootstrap_metric_ci(ya, yp, 'rmse')
            r2_ci   = bootstrap_metric_ci(ya, yp, 'r2')

            dist_results[arch][tgt] = {'rmse': rmse_ci, 'r2': r2_ci}
            print(f"   {arch:4s}/{tgt:28s}  "
                  f"RMSE {rmse_ci[0]:.3f} [{rmse_ci[1]:.3f},{rmse_ci[2]:.3f}]  "
                  f"R² {r2_ci[0]:.3f} [{r2_ci[1]:.3f},{r2_ci[2]:.3f}]")

    return dist_results


# ═══════════════════════════════════════════════════════════════════════════════
# 9.  ORIGINAL PLOT FUNCTIONS  (9A – 9J, unchanged)
# ═══════════════════════════════════════════════════════════════════════════════

# ── 9A: Actual vs Prediction + anomaly markers ────────────────────────────────
def plot_predictions(rba):
    for arch, tr in rba.items():
        cfg   = MODEL_CONFIGS[arch]
        valid = {t: r for t, r in tr.items() if r}
        if not valid: continue

        fig, axes = plt.subplots(3, 2, figsize=(22, 18))
        fig.patch.set_facecolor('white')
        axes = axes.flatten()
        fig.suptitle(f'{arch} — Actual vs Prediction  |  6 Targets',
                     fontsize=18, fontweight='bold')

        for idx, (tgt, res) in enumerate(valid.items()):
            ax = axes[idx]
            t  = pd.to_datetime(res['test_timestamps'])
            a, p = res['y_test_orig'], res['y_pred']

            ax.plot(t, a, color='#4361ee', lw=2,    label='Actual',     alpha=0.9)
            ax.plot(t, p, color=cfg['color'], lw=2,  label='Prediction',
                    ls='--', alpha=0.9)
            ax.fill_between(t, a, p, alpha=0.15, color=cfg['color'])

            fl = res['anomaly']['flags']
            ax.scatter(t[fl], a[fl], color='red', s=22, zorder=5,
                       label=f"Anomaly ({fl.sum()})")

            ax.set_title(
                f"{tgt}\nR²={res['r2']:.3f}  "
                f"nRMSE={res['nrmse']:.1f}%  "
                f"SeqAcc={res['seq_acc']['seq_score']:.1f}%",
                fontsize=9, fontweight='bold')
            ax.xaxis.set_major_formatter(mdates.DateFormatter('%d-%b\n%H:%M'))
            ax.legend(fontsize=8); ax.grid(True, alpha=0.4)

        plt.tight_layout()
        plt.savefig(f'/content/{arch}_predictions.png', dpi=150, bbox_inches='tight')
        plt.show()
        print(f"✅ Saved: {arch}_predictions.png")


# ── 9B: Permutation importance ────────────────────────────────────────────────
def plot_perm_xai(rba, top_n=10):
    for tgt in TARGETS:
        avail = {a: rba[a].get(tgt) for a in MODEL_CONFIGS if rba[a].get(tgt)}
        if not avail: continue

        n = len(avail)
        fig, axes = plt.subplots(1, n, figsize=(8*n, 6))
        if n == 1: axes = [axes]
        fig.suptitle(f'Permutation Feature Importance  |  {tgt}',
                     fontsize=14, fontweight='bold')

        for ax, (arch, res) in zip(axes, avail.items()):
            imp   = res['perm_imp']
            names = np.array(res['feature_names'])
            order = np.argsort(imp)[::-1][:top_n]
            bars  = ax.barh(names[order][::-1], imp[order][::-1],
                            color=MODEL_CONFIGS[arch]['color'], alpha=0.85)
            for b in bars:
                w = b.get_width()
                ax.text(w+0.002, b.get_y()+b.get_height()/2,
                        f'{w:.3f}', va='center', fontsize=8)
            ax.set_title(f'{arch}  (top {top_n})', fontweight='bold')
            ax.set_xlabel('Normalised ΔMSE')
            ax.grid(axis='x', alpha=0.4)

        plt.tight_layout()
        safe = tgt.replace(' ','_').replace('/','_').replace('%','pct')
        plt.savefig(f'/content/Perm_XAI_{safe}.png', dpi=150, bbox_inches='tight')
        plt.show()
        print(f"✅ Saved: Perm_XAI_{safe}.png")


# ── 9C: SHAP Bar ──────────────────────────────────────────────────────────────
def plot_shap_bar(rba, top_n=SHAP_TOP_N):
    for tgt in TARGETS:
        avail = {a: rba[a].get(tgt) for a in MODEL_CONFIGS
                 if rba[a].get(tgt) and rba[a][tgt].get('shap_2d') is not None}
        if not avail:
            print(f"   ⚠️  SHAP bar: no data for {tgt}"); continue

        n = len(avail)
        fig, axes = plt.subplots(1, n, figsize=(8*n, 6))
        if n == 1: axes = [axes]
        fig.suptitle(f'SHAP  mean |value|  — Feature Importance  |  {tgt}',
                     fontsize=14, fontweight='bold')

        for ax, (arch, res) in zip(axes, avail.items()):
            sv2   = res['shap_2d']
            names = np.array(res['feature_names'])
            m_abs = sv2.mean(axis=0)
            order = np.argsort(m_abs)[::-1][:top_n]

            pal  = plt.cm.RdYlGn_r(np.linspace(0.1, 0.9, top_n))
            bars = ax.barh(names[order][::-1], m_abs[order][::-1],
                           color=pal[::-1], alpha=0.9, edgecolor='white')
            for b in bars:
                w = b.get_width()
                ax.text(w+2e-4, b.get_y()+b.get_height()/2,
                        f'{w:.4f}', va='center', fontsize=8)

            ax.set_title(arch, fontweight='bold',
                         color=MODEL_CONFIGS[arch]['color'])
            ax.set_xlabel('mean(|SHAP value|)')
            ax.axvline(0, color='black', lw=0.8)
            ax.grid(axis='x', alpha=0.35)

        plt.tight_layout()
        safe = tgt.replace(' ','_').replace('/','_').replace('%','pct')
        plt.savefig(f'/content/SHAP_bar_{safe}.png', dpi=150, bbox_inches='tight')
        plt.show()
        print(f"✅ Saved: SHAP_bar_{safe}.png")


# ── 9D: SHAP Beeswarm ────────────────────────────────────────────────────────
def plot_shap_beeswarm(rba, top_n=SHAP_TOP_N):
    for arch in MODEL_CONFIGS:
        for tgt in TARGETS:
            res = rba[arch].get(tgt)
            if res is None or res.get('shap_3d') is None: continue

            sv3    = res['shap_3d']
            X_2d   = res['X_exp_2d']
            names  = np.array(res['feature_names'])

            sv_sig = sv3.mean(axis=1)
            m_abs  = np.abs(sv_sig).mean(axis=0)
            order  = np.argsort(m_abs)[::-1][:top_n]
            ordr   = order[::-1]
            nf     = len(ordr)

            fig, ax = plt.subplots(figsize=(10, max(5, nf*0.55)))
            fig.patch.set_facecolor('white')
            cmap = plt.cm.coolwarm

            for ri, fi in enumerate(ordr):
                sv_f  = sv_sig[:, fi]
                xf    = X_2d[:, fi]
                xf_n  = (xf - xf.min()) / (np.ptp(xf) + 1e-9)
                colors = cmap(xf_n)
                jitter = np.random.uniform(-0.3, 0.3, size=len(sv_f))
                ax.scatter(sv_f, ri + jitter, c=colors, s=12,
                           alpha=0.7, linewidths=0, zorder=3)

            ax.set_yticks(range(nf))
            ax.set_yticklabels(names[ordr], fontsize=9)
            ax.axvline(0, color='black', lw=1, zorder=4)
            ax.set_xlabel('SHAP value  (impact on model output)', fontsize=10)
            ax.set_title(f'SHAP Beeswarm  |  {arch}  —  {tgt}',
                         fontsize=12, fontweight='bold',
                         color=MODEL_CONFIGS[arch]['color'])
            ax.grid(axis='x', alpha=0.35)

            sm = plt.cm.ScalarMappable(cmap=cmap,
                                        norm=plt.Normalize(vmin=0, vmax=1))
            sm.set_array([])
            cb = plt.colorbar(sm, ax=ax, shrink=0.5, pad=0.01)
            cb.set_label('Feature value\n(low → high)', fontsize=8)

            plt.tight_layout()
            safe_t = tgt.replace(' ','_').replace('/','_').replace('%','pct')
            fname  = f'/content/SHAP_beeswarm_{arch}_{safe_t}.png'
            plt.savefig(fname, dpi=150, bbox_inches='tight')
            plt.show()
            print(f"✅ Saved: SHAP_beeswarm_{arch}_{safe_t}.png")


# ── 9E: SHAP Temporal Heatmap ─────────────────────────────────────────────────
def plot_shap_temporal(rba, top_n=8):
    for arch in MODEL_CONFIGS:
        for tgt in TARGETS:
            res = rba[arch].get(tgt)
            if res is None or res.get('shap_3d') is None: continue

            sv3   = res['shap_3d']
            names = np.array(res['feature_names'])

            heat   = np.abs(sv3).mean(axis=0)
            top_i  = np.argsort(heat.mean(axis=0))[::-1][:top_n]
            heat_t = heat[:, top_i]

            fig, ax = plt.subplots(figsize=(14, max(4, top_n*0.6)))
            fig.patch.set_facecolor('white')
            im = ax.imshow(heat_t.T, aspect='auto', cmap='YlOrRd', origin='lower')

            ax.set_yticks(range(top_n))
            ax.set_yticklabels(names[top_i], fontsize=9)
            ax.set_xlabel(
                'Sequence step  (0 = oldest  →  right = most recent)',
                fontsize=10)
            ax.set_title(
                f'SHAP Temporal Heatmap  |  {arch}  —  {tgt}\n'
                f'mean |SHAP| per timestep × feature  '
                f'(darker = more influential lag)',
                fontsize=11, fontweight='bold',
                color=MODEL_CONFIGS[arch]['color'])
            plt.colorbar(im, ax=ax, label='mean |SHAP|', shrink=0.8)
            plt.tight_layout()

            safe_t = tgt.replace(' ','_').replace('/','_').replace('%','pct')
            fname  = f'/content/SHAP_temporal_{arch}_{safe_t}.png'
            plt.savefig(fname, dpi=150, bbox_inches='tight')
            plt.show()
            print(f"✅ Saved: SHAP_temporal_{arch}_{safe_t}.png")


# ── 9F: SHAP Cross-Model ─────────────────────────────────────────────────────
def plot_shap_cross_model(rba, top_n=10):
    arch_list = list(MODEL_CONFIGS.keys())
    for tgt in TARGETS:
        avail = {a: rba[a].get(tgt) for a in arch_list
                 if rba[a].get(tgt) and rba[a][tgt].get('shap_2d') is not None}
        if not avail: continue

        union_mean = {}
        for arch, res in avail.items():
            mv = res['shap_2d'].mean(axis=0)
            for i, nm in enumerate(res['feature_names']):
                union_mean[nm] = union_mean.get(nm, 0) + mv[i]

        top_feats = sorted(union_mean, key=union_mean.get, reverse=True)[:top_n]

        fig, ax = plt.subplots(figsize=(14, 6))
        fig.patch.set_facecolor('white')
        x   = np.arange(len(top_feats))
        na  = len(avail)
        w   = 0.8 / na
        off = np.linspace(-(na-1)*w/2, (na-1)*w/2, na)

        for i, (arch, res) in enumerate(avail.items()):
            mv   = res['shap_2d'].mean(axis=0)
            fn   = res['feature_names']
            vals = [mv[fn.index(f)] if f in fn else 0 for f in top_feats]
            bars = ax.bar(x + off[i], vals, w,
                          label=arch, color=MODEL_CONFIGS[arch]['color'],
                          alpha=0.85, edgecolor='white')
            for b in bars:
                h = b.get_height()
                if h > 0:
                    ax.text(b.get_x()+b.get_width()/2, h+3e-4,
                            f'{h:.3f}', ha='center', va='bottom', fontsize=6.5)

        ax.set_xticks(x)
        ax.set_xticklabels(top_feats, rotation=35, ha='right', fontsize=9)
        ax.set_ylabel('mean |SHAP value|')
        ax.set_title(f'SHAP Cross-Model Comparison  |  {tgt}',
                     fontsize=13, fontweight='bold')
        ax.legend(fontsize=10); ax.grid(axis='y', alpha=0.35)
        plt.tight_layout()

        safe_t = tgt.replace(' ','_').replace('/','_').replace('%','pct')
        plt.savefig(f'/content/SHAP_crossmodel_{safe_t}.png', dpi=150, bbox_inches='tight')
        plt.show()
        print(f"✅ Saved: SHAP_crossmodel_{safe_t}.png")


# ── 9G: Anomaly dashboard ─────────────────────────────────────────────────────
def plot_anomaly_dashboard(rba):
    rows, cols = len(TARGETS), len(MODEL_CONFIGS)
    fig, axes  = plt.subplots(rows, cols, figsize=(7*cols, 4*rows))
    fig.patch.set_facecolor('white')
    fig.suptitle('Anomaly Score Dashboard  |  Z-score Distribution',
                 fontsize=16, fontweight='bold', y=1.01)

    arch_list = list(MODEL_CONFIGS.keys())
    for r, tgt in enumerate(TARGETS):
        for c, arch in enumerate(arch_list):
            ax  = axes[r][c]
            res = rba[arch].get(tgt)
            if res is None:
                ax.set_visible(False); continue

            z     = res['anomaly']['z_scores']
            flags = res['anomaly']['flags']
            ax.hist(z[~flags], bins=30, color=MODEL_CONFIGS[arch]['color'],
                    alpha=0.7, label='Normal')
            if flags.sum() > 0:
                ax.hist(z[flags], bins=10, color='red', alpha=0.85,
                        label=f"Anomaly ({flags.sum()})")
            ax.axvline(-ANOMALY_THRESH, color='red', ls='--', lw=1.2)
            ax.axvline( ANOMALY_THRESH, color='red', ls='--', lw=1.2)
            ax.set_title(
                f'{arch} | {tgt}\n'
                f'Rate={res["anomaly"]["anomaly_rate"]:.1f}%  '
                f'AvgZ={res["anomaly"]["mean_abs_zscore"]:.2f}',
                fontsize=8, fontweight='bold')
            ax.legend(fontsize=7); ax.grid(True, alpha=0.3)
            if c == 0: ax.set_ylabel('Count', fontsize=8)

    plt.tight_layout()
    plt.savefig('/content/anomaly_dashboard.png', dpi=150, bbox_inches='tight')
    plt.show()
    print("✅ Saved: anomaly_dashboard.png")


# ── 9H: Sequence Accuracy ─────────────────────────────────────────────────────
def plot_seq_accuracy(rba):
    arch_list = list(MODEL_CONFIGS.keys())
    fig, axes = plt.subplots(2, 3, figsize=(21, 10))
    fig.patch.set_facecolor('white')
    fig.suptitle('Sequence Prediction Accuracy  |  TBA & DA per Target',
                 fontsize=16, fontweight='bold')
    axes = axes.flatten()

    for idx, tgt in enumerate(TARGETS):
        ax = axes[idx]
        x  = np.arange(len(arch_list))
        w  = 0.25

        def _get(a, key):
            r = rba[a].get(tgt)
            return r['seq_acc'][key] if r else 0

        tba_v = [_get(a, 'tba') for a in arch_list]
        da_v  = [_get(a, 'da')  for a in arch_list]

        b1 = ax.bar(x-w/2, tba_v, w, label='TBA (±10%)',
                    color=[MODEL_CONFIGS[a]['color'] for a in arch_list], alpha=0.85)
        b2 = ax.bar(x+w/2, da_v, w, label='Directional Acc',
                    color=[MODEL_CONFIGS[a]['color'] for a in arch_list],
                    alpha=0.45, hatch='//')
        ax.set_xticks(x); ax.set_xticklabels(arch_list, fontweight='bold')
        ax.set_ylim(0, 115)
        ax.axhline(80, color='green', ls='--', lw=1, label='80% target')
        ax.set_title(tgt, fontweight='bold')
        ax.set_ylabel('Accuracy (%)')
        ax.legend(fontsize=7); ax.grid(axis='y', alpha=0.4)
        for b in list(b1)+list(b2):
            h = b.get_height()
            if h > 0:
                ax.text(b.get_x()+b.get_width()/2, h+0.5,
                        f'{h:.1f}', ha='center', fontsize=7)

    plt.tight_layout()
    plt.savefig('/content/seq_accuracy.png', dpi=150, bbox_inches='tight')
    plt.show()
    print("✅ Saved: seq_accuracy.png")


# ── 9I: Model Selection Dashboard ─────────────────────────────────────────────
def plot_model_selection_dashboard(rba):
    arch_list = list(MODEL_CONFIGS.keys())
    tgt_list  = TARGETS

    records = []
    for arch in arch_list:
        for tgt in tgt_list:
            res = rba[arch].get(tgt)
            if res is None:
                records.append({'arch': arch, 'target': tgt,
                                 'nRMSE': np.nan, 'R2': np.nan,
                                 'AnomalyRate': np.nan, 'SeqAcc': np.nan,
                                 'Composite': np.nan})
            else:
                records.append({'arch': arch, 'target': tgt,
                                 'nRMSE':       res['nrmse'],
                                 'R2':          res['r2'],
                                 'AnomalyRate': res['anomaly']['anomaly_rate'],
                                 'SeqAcc':      res['seq_acc']['seq_score'],
                                 'Composite':   composite(res)})
    summary = pd.DataFrame(records)

    fig = plt.figure(figsize=(28, 24))
    fig.patch.set_facecolor('white')
    gs  = gridspec.GridSpec(3, 2, figure=fig, hspace=0.45, wspace=0.35)
    ax_h = fig.add_subplot(gs[0, :])
    ax_n = fig.add_subplot(gs[1, 0])
    ax_r = fig.add_subplot(gs[1, 1])
    ax_a = fig.add_subplot(gs[2, 0])
    ax_s = fig.add_subplot(gs[2, 1])
    fig.suptitle('Model Selection Dashboard  |  RNN vs LSTM vs GRU',
                 fontsize=20, fontweight='bold', y=1.01)

    pv = summary.pivot(index='arch', columns='target', values='Composite')
    sns.heatmap(pv, ax=ax_h, annot=True, fmt='.1f', cmap='RdYlGn_r',
                linewidths=0.6, linecolor='white',
                cbar_kws={'label': 'Composite Score (lower = better)'})
    ax_h.set_title(
        'Composite Heatmap  '
        '(nRMSE×0.35 + R²pen×0.25 + Anomaly×0.25 + SeqPen×0.15)',
        fontweight='bold', fontsize=11)
    ax_h.set_xlabel(''); ax_h.set_ylabel('Architecture', fontweight='bold')
    ax_h.tick_params(axis='x', rotation=30)

    x    = np.arange(len(tgt_list))
    w    = 0.25
    offs = [-w, 0, w]
    xlabels = [t.replace(' (%)','\n(%)').replace('Efficiency ','Eff.')
               for t in tgt_list]

    for ax, metric, ylabel in [
        (ax_n, 'nRMSE',       'nRMSE (%)'),
        (ax_r, 'R2',          'R² Score'),
        (ax_a, 'AnomalyRate', 'Anomaly Rate (%)'),
        (ax_s, 'SeqAcc',      'Seq. Acc. (%)'),
    ]:
        for i, arch in enumerate(arch_list):
            vals = [summary[(summary.arch == arch) &
                            (summary.target == t)][metric].values[0]
                    for t in tgt_list]
            vals = np.array(vals, dtype=float)
            bars = ax.bar(x+offs[i], vals, w, label=arch,
                          color=MODEL_CONFIGS[arch]['color'], alpha=0.85)
            for b in bars:
                h = b.get_height()
                if np.isfinite(h) and h > 0:
                    ax.text(b.get_x()+b.get_width()/2, h+0.3,
                            f'{h:.1f}', ha='center', fontsize=6.5)
        ax.set_xticks(x); ax.set_xticklabels(xlabels, fontsize=8)
        ax.set_ylabel(ylabel, fontsize=9)
        ax.set_title(ylabel, fontweight='bold')
        ax.legend(fontsize=8); ax.grid(axis='y', alpha=0.35)

    plt.savefig('/content/model_selection_dashboard.png', dpi=150, bbox_inches='tight')
    plt.show()
    print("✅ Saved: model_selection_dashboard.png")
    return summary


# ── 9J: SHAP + Metrics joint dashboard ────────────────────────────────────────
def plot_shap_metrics_joint(rba, summary):
    arch_list = list(MODEL_CONFIGS.keys())
    tgt_list  = TARGETS

    fig = plt.figure(figsize=(30, 15))
    fig.patch.set_facecolor('white')
    gs  = gridspec.GridSpec(2, 4, figure=fig, hspace=0.55, wspace=0.42)
    fig.suptitle('SHAP + Metrics Joint Summary Dashboard',
                 fontsize=18, fontweight='bold', y=1.02)

    metric_cfg = [
        ('nRMSE',       'nRMSE (%)',       'YlOrRd'),
        ('R2',          'R² Score',         'YlGn'),
        ('AnomalyRate', 'Anomaly Rate (%)', 'OrRd'),
        ('SeqAcc',      'Seq. Acc. (%)',    'BuGn'),
    ]

    for col, (metric, title, cmap) in enumerate(metric_cfg):
        ax = fig.add_subplot(gs[0, col])
        pv = summary.pivot(index='arch', columns='target', values=metric)
        sns.heatmap(pv, ax=ax, annot=True, fmt='.1f', cmap=cmap,
                    linewidths=0.5, linecolor='white', cbar=False)
        ax.set_title(title, fontweight='bold', fontsize=11)
        ax.set_xlabel(''); ax.set_ylabel('Arch', fontsize=9)
        ax.tick_params(axis='x', rotation=35, labelsize=8)

    for col, arch in enumerate(arch_list):
        ax = fig.add_subplot(gs[1, col])
        combined = {}
        for tgt in tgt_list:
            res = rba[arch].get(tgt)
            if res is None or res.get('shap_2d') is None: continue
            mv = res['shap_2d'].mean(axis=0)
            for i, nm in enumerate(res['feature_names']):
                combined.setdefault(nm, []).append(mv[i])

        if not combined:
            ax.text(0.5, 0.5, 'SHAP N/A', ha='center', transform=ax.transAxes)
            continue

        feat_mean = {k: np.mean(v) for k, v in combined.items()}
        top_f     = sorted(feat_mean, key=feat_mean.get, reverse=True)[:10]
        vals      = [feat_mean[f] for f in top_f]
        pal       = plt.cm.Blues(np.linspace(0.35, 0.9, len(top_f)))
        ax.barh(top_f[::-1], vals[::-1], color=pal, alpha=0.9, edgecolor='white')
        ax.set_title(f'SHAP Overall\n{arch}', fontweight='bold',
                     color=MODEL_CONFIGS[arch]['color'], fontsize=11)
        ax.set_xlabel('mean |SHAP| (all targets)')
        ax.grid(axis='x', alpha=0.35)

    plt.savefig('/content/SHAP_metrics_joint.png', dpi=150, bbox_inches='tight')
    plt.show()
    print("✅ Saved: SHAP_metrics_joint.png")


# ═══════════════════════════════════════════════════════════════════════════════
#  [NEW v2]  9K — EPOCHS EFFECT PLOT  (GRU, 4-panel)
# ═══════════════════════════════════════════════════════════════════════════════

def plot_epochs_effect(epoch_data, full_history):
    """
    9K: 4-panel figure.
      [0,0] Full training/validation loss curve (all epochs, log scale).
      [0,1] Test nRMSE at checkpoint epochs — shows when generalisation peaks.
      [1,0] Test R²   at checkpoint epochs.
      [1,1] Test MAPE at checkpoint epochs.
    Red star = best value per metric.  Red dashed line = optimal epoch.
    """
    if not epoch_data or not full_history:
        print("   ⚠️  No epochs effect data to plot"); return

    color = MODEL_CONFIGS['GRU']['color']
    epochs_all = np.arange(1, len(full_history['loss']) + 1)

    fig, axes = plt.subplots(2, 2, figsize=(16, 10))
    fig.patch.set_facecolor('white')
    fig.suptitle(
        f'Epochs Effect Analysis  |  GRU 1-Layer  '
        f'(seq={GRU_EPOCH_SEQ}, units={GRU_EPOCH_UNITS})\n'
        f'Target: {EPOCH_ANALYSIS_TGT}  '
        f'[Checkpoints: {EPOCH_CHECKPOINTS}]',
        fontsize=13, fontweight='bold'
    )
    axes = axes.flatten()

    # ── Panel 0: Full loss curves ──────────────────────────────────────────
    ax = axes[0]
    ax.plot(epochs_all, full_history['loss'],
            color='#4361ee', lw=2, label='Train loss', alpha=0.9)
    ax.plot(epochs_all, full_history['val_loss'],
            color=color, lw=2, ls='--', label='Val loss', alpha=0.85)
    best_ep  = int(np.argmin(full_history['val_loss'])) + 1
    best_val = float(np.min(full_history['val_loss']))
    ax.axvline(best_ep, color='red', ls=':', lw=1.8, alpha=0.75)
    ax.scatter([best_ep], [best_val], color='red', s=90, zorder=5,
               label=f'Best val @ ep {best_ep}')
    ax.set_xlabel('Epoch'); ax.set_ylabel('MSE Loss (scaled)')
    ax.set_title('Training & Validation Loss Curve', fontweight='bold')
    ax.set_yscale('log')
    ax.legend(fontsize=9); ax.grid(True, alpha=0.4)

    # Helper: parse checkpoint records
    ep_list   = [r['epoch']  for r in epoch_data]
    nrmse_list = [r['nrmse'] for r in epoch_data]
    r2_list   = [r['r2']    for r in epoch_data]
    mape_list = [r['mape']  for r in epoch_data]

    def _annotate_best(ax_in, ep_l, val_l, best_fn, label, col):
        bi = best_fn(val_l)
        ax_in.scatter([ep_l[bi]], [val_l[bi]], color='red', s=130,
                      zorder=6, marker='*',
                      label=f'Best @ ep {ep_l[bi]}  ({val_l[bi]:.3f})')
        ax_in.axvline(ep_l[bi], color='red', ls='--', lw=1.2, alpha=0.5)
        ax_in.axhline(val_l[bi], color='grey', ls=':', lw=1, alpha=0.5)

    # ── Panel 1: nRMSE ────────────────────────────────────────────────────
    ax = axes[1]
    ax.plot(ep_list, nrmse_list, color=color, lw=2.5, marker='o', ms=7,
            label='Test nRMSE (%)')
    _annotate_best(ax, ep_list, nrmse_list, np.argmin, 'nRMSE', color)
    ax.set_xlabel('Epoch'); ax.set_ylabel('nRMSE (%)')
    ax.set_title('Test nRMSE vs Epochs', fontweight='bold')
    ax.legend(fontsize=9); ax.grid(True, alpha=0.4)

    # ── Panel 2: R² ───────────────────────────────────────────────────────
    ax = axes[2]
    ax.plot(ep_list, r2_list, color='#2a9d8f', lw=2.5, marker='s', ms=7,
            label='Test R²')
    _annotate_best(ax, ep_list, r2_list, np.argmax, 'R²', '#2a9d8f')
    ax.set_xlabel('Epoch'); ax.set_ylabel('R²')
    ax.set_title('Test R² vs Epochs', fontweight='bold')
    ax.legend(fontsize=9); ax.grid(True, alpha=0.4)

    # ── Panel 3: MAPE ─────────────────────────────────────────────────────
    ax = axes[3]
    ax.plot(ep_list, mape_list, color='#e63946', lw=2.5, marker='^', ms=7,
            label='Test MAPE (%)')
    _annotate_best(ax, ep_list, mape_list, np.argmin, 'MAPE', '#e63946')
    ax.set_xlabel('Epoch'); ax.set_ylabel('MAPE (%)')
    ax.set_title('Test MAPE vs Epochs', fontweight='bold')
    ax.legend(fontsize=9); ax.grid(True, alpha=0.4)

    plt.tight_layout()
    plt.savefig('/content/epochs_effect_GRU.png', dpi=150, bbox_inches='tight')
    plt.show()
    print("✅ Saved: epochs_effect_GRU.png")

    # Summary table
    print("\n   📊 Epochs Effect Summary (GRU):")
    print(f"      Best val-loss epoch : {best_ep}")
    bi_n = int(np.argmin(nrmse_list))
    bi_r = int(np.argmax(r2_list))
    bi_m = int(np.argmin(mape_list))
    print(f"      Best test nRMSE     : epoch {ep_list[bi_n]:3d}  →  {nrmse_list[bi_n]:.2f}%")
    print(f"      Best test R²        : epoch {ep_list[bi_r]:3d}  →  {r2_list[bi_r]:.4f}")
    print(f"      Best test MAPE      : epoch {ep_list[bi_m]:3d}  →  {mape_list[bi_m]:.2f}%")


# ═══════════════════════════════════════════════════════════════════════════════
#  [NEW v2]  9L — TIME-SERIES CV COMPARISON PLOT
# ═══════════════════════════════════════════════════════════════════════════════

def plot_cv_comparison(cv_results):
    """
    9L: 3-row × 3-col figure.
      Row 0: Blocked CV — mean nRMSE per target (bar), error bars = fold std,
             individual fold points overlaid.
      Row 1: Rolling-Origin CV — same layout.
      Row 2: Strategy difference (Blocked nRMSE − Rolling nRMSE) per arch:
             red bars = Blocked is worse; teal bars = Rolling is worse.

    Follows Roberts 2017 / Cerqueira 2019 (Blocked) and
    Schnaubelt 2020 / Maldonado 2021 (Rolling-Origin).
    """
    arch_list       = list(MODEL_CONFIGS.keys())
    x               = np.arange(len(TARGETS))
    xlabels         = [t.replace(' (%)', '\n(%)').replace('Efficiency ', 'Eff.')
                       for t in TARGETS]
    strategy_meta   = [
        ('blocked', 'Blocked Time-Split CV\n(Roberts 2017; Cerqueira 2019)', 0),
        ('rolling', 'Rolling-Origin CV\n(Schnaubelt 2020; Maldonado 2021)', 1),
    ]

    fig = plt.figure(figsize=(22, 19))
    fig.patch.set_facecolor('white')
    gs  = gridspec.GridSpec(3, len(arch_list), figure=fig,
                             hspace=0.55, wspace=0.38)
    fig.suptitle(
        'Time-Series Cross-Validation Comparison\n'
        'Bars = fold-mean ± fold-std  |  Dots = individual folds  '
        '|  nRMSE (%) lower = better',
        fontsize=14, fontweight='bold'
    )

    for strat, strat_label, row in strategy_meta:
        for col, arch in enumerate(arch_list):
            ax = fig.add_subplot(gs[row, col])

            means, stds = [], []
            for tgt in TARGETS:
                folds = cv_results[strat][arch].get(tgt, [])
                vals  = [f['nrmse'] for f in folds
                         if f and not np.isnan(f.get('nrmse', np.nan))]
                means.append(np.mean(vals) if vals else np.nan)
                stds.append(np.std(vals)   if len(vals) > 1 else 0)

            means = np.array(means, dtype=float)
            stds  = np.array(stds,  dtype=float)
            valid = ~np.isnan(means)

            ax.bar(x[valid], means[valid], 0.55,
                   color=MODEL_CONFIGS[arch]['color'], alpha=0.75)
            ax.errorbar(x[valid], means[valid], yerr=stds[valid],
                        fmt='none', color='#333333',
                        capsize=5, lw=2, zorder=5)

            # Overlay fold-level dots
            for fi, tgt in enumerate(TARGETS):
                folds = cv_results[strat][arch].get(tgt, [])
                pts   = [f['nrmse'] for f in folds
                         if f and not np.isnan(f.get('nrmse', np.nan))]
                if pts:
                    jitter = np.random.uniform(-0.12, 0.12, len(pts))
                    ax.scatter([fi + j for j in jitter], pts,
                               color='#222222', s=28, zorder=6, alpha=0.75)

            ax.set_xticks(x)
            ax.set_xticklabels(xlabels, fontsize=7)
            ax.set_ylabel('nRMSE (%)', fontsize=8)
            ax.set_title(f'{strat_label}\n{arch}',
                         fontweight='bold',
                         color=MODEL_CONFIGS[arch]['color'], fontsize=8.5)
            ax.axhline(20, color='green', ls='--', lw=1,
                       alpha=0.6, label='20% ref')
            ax.legend(fontsize=7)
            ax.grid(axis='y', alpha=0.35)

    # Row 2: strategy difference per arch
    for col, arch in enumerate(arch_list):
        ax = fig.add_subplot(gs[2, col])

        diffs = []
        for tgt in TARGETS:
            b_folds = cv_results['blocked'][arch].get(tgt, [])
            r_folds = cv_results['rolling'][arch].get(tgt, [])
            b_vals  = [f['nrmse'] for f in b_folds
                       if f and not np.isnan(f.get('nrmse', np.nan))]
            r_vals  = [f['nrmse'] for f in r_folds
                       if f and not np.isnan(f.get('nrmse', np.nan))]
            if b_vals and r_vals:
                diffs.append(np.mean(b_vals) - np.mean(r_vals))
            else:
                diffs.append(np.nan)

        diffs  = np.array(diffs, dtype=float)
        valid  = ~np.isnan(diffs)
        colors = ['#e63946' if d > 0 else '#2a9d8f'
                  for d in diffs[valid]]

        ax.bar(x[valid], diffs[valid], 0.55, color=colors, alpha=0.8)
        ax.axhline(0, color='black', lw=1.8)
        ax.set_xticks(x)
        ax.set_xticklabels(xlabels, fontsize=7)
        ax.set_ylabel('ΔnRMSE\n(Blocked − Rolling)', fontsize=8)
        ax.set_title(
            f'Strategy Diff — {arch}\n'
            '🔴 +ve = Blocked worse  |  🟢 −ve = Rolling worse',
            fontsize=8, fontweight='bold')
        ax.grid(axis='y', alpha=0.35)

        # Annotate each bar
        for xi, (d, v) in enumerate(zip(diffs[valid], x[valid])):
            ax.text(v, d + (0.2 if d >= 0 else -0.5),
                    f'{d:+.1f}', ha='center', fontsize=6.5)

    plt.savefig('/content/CV_comparison.png', dpi=150, bbox_inches='tight')
    plt.show()
    print("✅ Saved: CV_comparison.png")

    # Print CV summary table
    _print_cv_summary(cv_results)


def _print_cv_summary(cv_results):
    """Print a compact CV summary table to console."""
    arch_list = list(MODEL_CONFIGS.keys())
    print("\n" + "=" * 90)
    print("  CV SUMMARY  (mean nRMSE % across folds — lower = better)")
    print("=" * 90)
    hdr = f"{'Target':<28}" + "".join(
        f"{'Blocked-'+a:>14}" for a in arch_list
    ) + "".join(f"{'Rolling-'+a:>14}" for a in arch_list)
    print(hdr)
    print("-" * 90)
    for tgt in TARGETS:
        row = f"{tgt:<28}"
        for strat in ['blocked', 'rolling']:
            for arch in arch_list:
                folds = cv_results[strat][arch].get(tgt, [])
                vals  = [f['nrmse'] for f in folds
                         if f and not np.isnan(f.get('nrmse', np.nan))]
                row += f"{np.mean(vals):>14.2f}" if vals else f"{'N/A':>14}"
        print(row)
    print("=" * 90)


# ═══════════════════════════════════════════════════════════════════════════════
#  [NEW v2]  9M — AGGREGATE VALIDATION PLOT
# ═══════════════════════════════════════════════════════════════════════════════

def plot_aggregate_validation(agg_results):
    """
    9M: 2-row × 6-col figure.
      Row 0: RMSE at point / hourly / daily level per arch.
      Row 1: R²  at the same levels.

    Visualises how temporal aggregation reduces prediction error —
    key insight from Kennedy et al. 2023: the relevant metric must
    be measured at the inference level, not at the raw sample level.
    """
    arch_list = list(MODEL_CONFIGS.keys())
    n_tgt     = len(TARGETS)
    levels    = [('point', 'Point'),
                 ('hourly', 'Hourly'), ('daily', 'Daily')]

    fig, axes = plt.subplots(2, n_tgt, figsize=(26, 10))
    fig.patch.set_facecolor('white')
    fig.suptitle(
        'Aggregate Validation  |  Point → Hourly → Daily Error\n'
        '(Kennedy et al. 2023: validate at the aggregate inference level)',
        fontsize=13, fontweight='bold'
    )

    x_arch  = np.arange(len(arch_list))
    grp_w   = 0.8
    bar_w   = grp_w / len(levels)
    offsets = np.linspace(-(grp_w-bar_w)/2, (grp_w-bar_w)/2, len(levels))
    level_colors = ['#264653', '#2a9d8f', '#e9c46a']
    level_alphas = [0.95, 0.70, 0.45]

    for col, tgt in enumerate(TARGETS):
        for row, (metric, mlabel) in enumerate([('rmse', 'RMSE'),
                                                 ('r2',   'R²')]):
            ax = axes[row][col]
            any_data = False

            for li, (lvl, lvl_label) in enumerate(levels):
                vals = []
                for arch in arch_list:
                    d    = agg_results.get(arch, {}).get(tgt, {})
                    vals.append(d.get(f'{lvl}_{metric}', np.nan))
                vals = np.array(vals, dtype=float)
                valid = ~np.isnan(vals)
                if valid.any():
                    any_data = True
                    bars = ax.bar(
                        x_arch[valid] + offsets[li],
                        vals[valid], bar_w,
                        color=level_colors[li],
                        alpha=level_alphas[li],
                        label=lvl_label if col == 0 else '_'
                    )
                    for b in bars:
                        h = b.get_height()
                        if np.isfinite(h) and h > 0:
                            ax.text(b.get_x()+b.get_width()/2, h*1.015,
                                    f'{h:.2f}', ha='center',
                                    fontsize=6, va='bottom')

            ax.set_xticks(x_arch)
            ax.set_xticklabels(arch_list, fontsize=8)
            safe_t = tgt.replace(' (%)', '\n(%)').replace('Efficiency ', 'Eff.')
            ax.set_title(safe_t, fontsize=8, fontweight='bold')
            ax.set_ylabel(mlabel, fontsize=8)
            ax.grid(axis='y', alpha=0.35)

            if col == 0:
                ax.legend(fontsize=7, title='Agg. level')

            if not any_data:
                ax.text(0.5, 0.5, 'No data', ha='center',
                        transform=ax.transAxes, fontsize=9)

    plt.tight_layout()
    plt.savefig('/content/aggregate_validation.png', dpi=150, bbox_inches='tight')
    plt.show()
    print("✅ Saved: aggregate_validation.png")


# ═══════════════════════════════════════════════════════════════════════════════
#  [NEW v2]  9N — BOOTSTRAP CI / PERFORMANCE DISTRIBUTION PLOT
# ═══════════════════════════════════════════════════════════════════════════════

def plot_performance_distribution(dist_results):
    """
    9N: 2-row × 6-col figure showing bootstrapped 95% CI for RMSE and R².
    Bars = bootstrap mean; error bars = [lower, upper] of the CI.
    Each architecture colour-coded.

    Implements Lehmann et al. 2025:
    "Beyond point estimates — distributional uncertainty in forecasting."
    """
    arch_list = list(MODEL_CONFIGS.keys())
    x         = np.arange(len(arch_list))
    w         = 0.5

    fig, axes = plt.subplots(2, len(TARGETS), figsize=(26, 10))
    fig.patch.set_facecolor('white')
    fig.suptitle(
        f'Bootstrap {int(BOOTSTRAP_CI*100)}% CI  |  '
        f'Distributional Performance Uncertainty  (N={BOOTSTRAP_N} resamples)\n'
        '(Lehmann et al. 2025: beyond point estimates)',
        fontsize=13, fontweight='bold'
    )

    for col, tgt in enumerate(TARGETS):
        for row, (metric, ylabel) in enumerate([('rmse', 'RMSE'),
                                                 ('r2',   'R²')]):
            ax = axes[row][col]

            means, lowers, uppers, colors = [], [], [], []
            for arch in arch_list:
                ci = dist_results.get(arch, {}).get(tgt, {}).get(metric,
                                                                  (np.nan,)*3)
                means.append(ci[0]); lowers.append(ci[1]); uppers.append(ci[2])
                colors.append(MODEL_CONFIGS[arch]['color'])

            means  = np.array(means,  dtype=float)
            lowers = np.array(lowers, dtype=float)
            uppers = np.array(uppers, dtype=float)
            valid  = ~np.isnan(means)

            if not valid.any():
                ax.text(0.5, 0.5, 'No data', ha='center',
                        transform=ax.transAxes); continue

            bars = ax.bar(x[valid], means[valid], w,
                          color=[colors[i] for i in range(len(arch_list))
                                 if valid[i]],
                          alpha=0.78, zorder=3)

            # 95% CI error bars
            yerr = np.vstack([
                np.where(np.isnan(means - lowers), 0, means - lowers),
                np.where(np.isnan(uppers - means), 0, uppers - means),
            ])
            ax.errorbar(x[valid], means[valid],
                        yerr=[yerr[0, valid], yerr[1, valid]],
                        fmt='none', color='#111111',
                        capsize=6, lw=2.2, zorder=5,
                        label=f'{int(BOOTSTRAP_CI*100)}% CI')

            # Annotate bars
            for xi, m in zip(x[valid], means[valid]):
                ax.text(xi, m*1.01 if m > 0 else m - abs(m)*0.1,
                        f'{m:.3f}', ha='center', fontsize=6.5, va='bottom')

            ax.set_xticks(x)
            ax.set_xticklabels(arch_list, fontsize=8)
            safe_t = tgt.replace(' (%)', '\n(%)').replace('Efficiency ', 'Eff.')
            ax.set_title(safe_t, fontsize=8, fontweight='bold')
            ax.set_ylabel(ylabel, fontsize=8)
            ax.grid(axis='y', alpha=0.35)

            if col == 0:
                ax.legend(fontsize=7)

    plt.tight_layout()
    plt.savefig('/content/performance_distribution.png', dpi=150, bbox_inches='tight')
    plt.show()
    print("✅ Saved: performance_distribution.png")


# ─────────────────────────────────────────────────────────────────────────────
# 10.  WINNER TABLE
# ─────────────────────────────────────────────────────────────────────────────
def print_winner_table(summary):
    print("\n" + "=" * 92)
    print("  MODEL SELECTION — BEST ARCHITECTURE PER TARGET  (Composite score)")
    print("=" * 92)
    cw = [28, 6, 8, 6, 13, 10, 14]
    print(f"{'Target':<{cw[0]}} {'Arch':<{cw[1]}} {'nRMSE%':<{cw[2]}} "
          f"{'R²':<{cw[3]}} {'AnomalyRate%':<{cw[4]}} "
          f"{'SeqAcc%':<{cw[5]}} {'Composite':<{cw[6]}}")
    print("-" * 92)

    votes = {a: 0 for a in MODEL_CONFIGS}
    for tgt in TARGETS:
        sub = summary[summary.target == tgt].dropna(subset=['Composite'])
        if sub.empty: continue
        best = sub.loc[sub['Composite'].idxmin()]
        votes[best['arch']] += 1
        print(f"{tgt:<{cw[0]}} {best['arch']:<{cw[1]}} "
              f"{best['nRMSE']:<{cw[2]}.2f} {best['R2']:<{cw[3]}.3f} "
              f"{best['AnomalyRate']:<{cw[4]}.1f} "
              f"{best['SeqAcc']:<{cw[5]}.1f} "
              f"{best['Composite']:<{cw[6]}.2f}")

    best_overall = max(votes, key=votes.get)
    print("=" * 92)
    print(f"\n  OVERALL BEST → {best_overall}   |   Votes: {votes}")
    print("\n  Composite formula (lower = better):")
    print("    0.35 × nRMSE  +  0.25 × (1−R²)×100  "
          "+  0.25 × AnomalyRate  +  0.15 × (100−SeqAcc)")
    print("=" * 92)

    pd.set_option('display.float_format', '{:.3f}'.format)
    pd.set_option('display.max_columns', 20)
    pd.set_option('display.width', 130)
    print("\n  FULL TABLE:")
    print(summary.sort_values(['target', 'Composite']).to_string(index=False))


# ─────────────────────────────────────────────────────────────────────────────
# 11.  MAIN
# ─────────────────────────────────────────────────────────────────────────────
def main():
    print("\n📁 Step 1/9 — Data loading & preprocessing")
    df = load_hybrid_data()
    if len(df) < 200:
        print("❌ Insufficient data!"); return
    df = engineer_features(df,
                           system_capacity_wp=TRAIN_CFG['system_capacity_wp'],
                           panel_area_m2=TRAIN_CFG['panel_area_m2'])
    df = clean_data(df)

    print("\n🤖 Step 2/9 — Training 3 arch × 6 targets = 18 models")
    print("─" * 72)
    rba = {arch: {} for arch in MODEL_CONFIGS}
    for arch in MODEL_CONFIGS:
        cfg = MODEL_CONFIGS[arch]
        print(f"\n🔷 {arch}   seq={cfg['seq']}   units={cfg['units']}")
        for tgt in TARGETS:
            if tgt in df.columns:
                rba[arch][tgt] = train_model(df, tgt, arch)
            else:
                print(f"   ⚠️  Column '{tgt}' not found — skipping")
                rba[arch][tgt] = None

    print("\n📈 Step 3/9 — Prediction time-series plots")
    plot_predictions(rba)

    print("\n🔍 Step 4/9 — Permutation Feature Importance")
    plot_perm_xai(rba, top_n=10)

    print("\n🎯 Step 5/9 — SHAP XAI (bar + beeswarm + temporal + cross-model)")
    plot_shap_bar(rba)
    plot_shap_beeswarm(rba)
    plot_shap_temporal(rba)
    plot_shap_cross_model(rba)

    print("\n🔴 Step 5b/9 — Anomaly Dashboard + Sequence Accuracy")
    plot_anomaly_dashboard(rba)
    plot_seq_accuracy(rba)

    print("\n🏆 Step 6/9 — Model Selection Dashboard + Winner Table")
    summary = plot_model_selection_dashboard(rba)
    plot_shap_metrics_joint(rba, summary)
    print_winner_table(summary)
    summary.to_csv('/content/model_comparison_results.csv', index=False)
    print("💾 Saved → /content/model_comparison_results.csv")

    # ── [NEW v2] Step 7: Epochs Effect Analysis ────────────────────────────
    print("\n" + "=" * 72)
    print("  [NEW v2]  Step 7/9 — Epochs Effect Analysis (GRU 1-layer)")
    print("=" * 72)
    epoch_data, full_history = analyze_epochs_effect(df)
    if epoch_data:
        plot_epochs_effect(epoch_data, full_history)

    # ── [NEW v2] Step 8: Time-Series Cross-Validation ─────────────────────
    print("\n" + "=" * 72)
    print("  [NEW v2]  Step 8/9 — Time-Series Cross-Validation")
    print(f"           Blocked (Roberts 2017)  +  Rolling-Origin (Schnaubelt 2020)")
    print(f"           {CV_N_FOLDS} folds  |  no random shuffle  |  temporal order preserved")
    print("=" * 72)
    cv_results = run_timeseries_cv(df, n_folds=CV_N_FOLDS)
    plot_cv_comparison(cv_results)

    # Save CV results
    cv_rows = []
    for strat in ['blocked', 'rolling']:
        for arch in MODEL_CONFIGS:
            for tgt in TARGETS:
                for fold in cv_results[strat][arch].get(tgt, []):
                    cv_rows.append({
                        'strategy': strat, 'arch': arch, 'target': tgt,
                        **fold
                    })
    if cv_rows:
        pd.DataFrame(cv_rows).to_csv('/content/cv_results.csv', index=False)
        print("💾 Saved → /content/cv_results.csv")

    # ── [NEW v2] Step 9: Aggregate Validation + Bootstrap CI ──────────────
    print("\n" + "=" * 72)
    print("  [NEW v2]  Step 9/9 — Aggregate Validation + Bootstrap CI")
    print("           Aggregate: Kennedy et al. 2023")
    print("           Bootstrap: Lehmann et al. 2025")
    print("=" * 72)
    agg_results  = aggregate_validate(rba)
    dist_results = compute_performance_distributions(rba)
    plot_aggregate_validation(agg_results)
    plot_performance_distribution(dist_results)

    print("\n" + "=" * 80)
    print("✅  ALL DONE  |  v2 complete")
    print("   Outputs saved to /content/")
    print("   New plots: epochs_effect_GRU.png | CV_comparison.png |")
    print("              aggregate_validation.png | performance_distribution.png")
    print("=" * 80)


if __name__ == '__main__':
    main()
```
